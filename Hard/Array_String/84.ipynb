```python
from typing import List

# DOes not work because too BRUTE FORCE
# Brute Force
class Solution:
    def largestRectangleArea(self, heights: List[int]) -> int:
        

        # creating the cache
        cacheMatrix = [[-1]*len(heights) for _ in range(len(heights))]

        def helperFn(index_1, index_2, cache):

            # if the index is further than the len, return 0
            if index_1==len(heights) or index_2==len(heights):
                return 0 

            # if first element area is 0
            if index_1>index_2:
                return 0
            
            # checking if in cache
            if cache[index_1][index_2]!=-1:
                return cache[index_1][index_2]

            # calculating the area
            new_array = heights[index_1:index_2+1]            

            area = min(new_array)*(index_2-index_1+1)

            # have a choice to move either one, or both
            answer = max(area,
                        helperFn(index_1+1, index_2, cache),
                        helperFn(index_1, index_2+1, cache),
                        helperFn(index_1+1, index_2+1, cache),
            )

            # saving the answer
            cache[index_1][index_2] = answer

            return cache[index_1][index_2]

        return helperFn(0,0, cacheMatrix)
```

```python
class Solution:
    def largestRectangleArea(self, heights: List[int]) -> int:
        
        # for every rectangle, move to the left and right and see what is the biggest area you can create
        maximum_area = 0 
        max_index = len(heights)

        # function that calcs the area
        def helperFn(rect_index, rect_height):            

            # calc left_side, find the minimum height then area
            left_array = heights[0:rect_index]
            left_array = left_array[::-1]

            rect_length = 0

            for i in left_array:
                if i >= rect_height:
                    rect_length+=1
                else:
                    break
            
            ## Right calculations
            # include the current rect this time
            right_array = heights[rect_index:max_index]

            for i in right_array:
                if i >= rect_height:
                    rect_length+=1
                else:
                    break
            
            rect_area = rect_length*rect_height

            # how far left can move and how far right can move
            return rect_area

        for index, rectangle in enumerate(heights):
            maximum_area = max(maximum_area, helperFn(index, rectangle))
        
        return maximum_area

# time complexity of O(n^2)
# space complexity of O(n)
```
