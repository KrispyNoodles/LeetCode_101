### Creating with CONSTRAINTS

The 'NOT NULL' CONSTRAINT is a constraint used in SQL.

```sql
CREATE TABLE table_name(
    price INTEGER NOT NULL,
    name TEXT NOT NULL DEFAULT 'something'
);
```

when there isnt a default placed, it will raise an error when the user tries to create without a default value.

### The 'UNIQUE' CONSTRAINT

This means that whenever there is an id added into the table it has to be unique

```sql
CREATE TABLE students(
  id INTEGER UNIQUE NOT NULL,
  name TEXT NOT NULL,
  age INTEGER NOT NULL
)
```

### PRIMARY KEY

A PRIMARY KEY IS SPECIAL COLUMN within the table and it must be UNIQUE and cannot contain a NULL VALUE

```sql
CREATE TABLE users(
  id INTEGER PRIMARY KEY,
  username TEXT
)
```

### FOREIGN KEY

A FOREIGN KEY IS IS A COLUMN that references to a column in another table. 
This means then whenever a value is inserted into the owner_id of videos, that same value must be found in the column of the users id, if not it will fail

(IT IS NOT STATED "FOREIGN KEY" but it is called "REFERENCES")

```sql
CREATE TABLE users (
    id INTEGER PRIMARY KEY,
    name TEXT
)

CREATE TABLE videos (
    id INTEGER PRIMARY KEY,
    title TEXT,
    owner_id INTEGER REFERENCES users(id)
)
```

### COMPOSITE PRIMARY KEY

A COMPOSITE PRIMARY KEY is made up of different columns in the same table, which means then whenever a value is inserted into the owner_id of videos, that same value must be found in the column of the users id, if not it will fail

(IT IS NOT STATED "FOREIGN KEY" but it is called "REFERENCES")

```sql
CREATE TABLE orders(
  order_id INTEGER,
  product_id INTEGER,
  quantity INTEGER,
  PRIMARY KEY (order_id, product_id)
);


### CONSTRAINT CHECK 

```sql
variable_name INTEGER CHECK (variable_name IN ('string_1', 'string_2'))
