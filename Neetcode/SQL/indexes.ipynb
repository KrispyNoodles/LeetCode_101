### Indexes

Database use indexes to speed up data retrieval. An index is like a table of contents for data stored with each table in a database.

It can be created by using the `CREATE INDEX` keyword

```sql
-- it is common to have an index after the column it is indexing, followed by a '_idx' or '_index'
-- users(name) follows the format of table_name(table_column)
CREATE INDEX name_idx ON users(name);
```

This means that queries that filter data based on the `name` column will be alot faster because of the index
```sql
SELECT * FROM users WHERE name = 'John';


### Types of Indexes

**B-Tree Index**:
- The most common type of index and suitable for most data type
- Useful for range queries (between...) and equality checks
- Similar to Binary Trees so they can perform lookups in O(h) time, h refers to the height of the tree

```sql
CREATE INDEX idx_name ON pokemon(name);

--same as
CREATE INDEX idx_name ON pokemon USING btree(name);
```

**Hash Index**:
- Uses a hsash function to quickly locate data
- Efficient for equality checks on small data types
- Similar to Hash Tables so they can perform lookups in O(1) time 

In theory, hash indexes are faster but most cases B-Trees perform about the same and are better for range queries. Therefore it is used as the default index type in PostgreSQL.

```sql
--same as
CREATE INDEX idx_name ON pokemon USING hash(name);
```


### Viewing Indexes

You can view the indexes on a table by querying `pg_indexes` view

```sql
SELECT *
FROM pg_indexes
WHERE tablename = 'employees';
```

By default, all primary key will have a index created when the table is created.


We can create additional indexes even after the table is created. If alot of data is already inserted into the table, the index creation may take time. While indexes speed up read queries, they slow down write queries. Because every time a write query is executed, the index must be re-built to reflect the new data. Therefore, if we have multiple index on a table, the write-query will be slower
