# Caching

The CPU cache is the computer's speediest memory, allowing the CPU to perform read and write operations slightly faster than on RAM or disk. Caching is simply the process of making copies of data. In a single computer, this can mean copying data from RAM into the CPU's cache memory (data on disk is first loaded into RAM before being cached by the CPU). Even though this data is already present in RAM and disk, duplicating it in the cache allows us to read and write data at much quicker rates. Even though the cache outperforms disk and RAM in speed, its capacity is restricted and usually in the kilobytes or megabytes. As a result, the CPU (not the OS) automatically decides what data to store in the cache.

## Real example from neetcode
How does a browser decide which files to cache on disk? It is usually the static elements that get cached. This includes images or pages whose content remains unchanged no matter how many times they are refreshed. For example, on https://neetcode.io, if we navigate to the network tab, we can find a file that looks like main.cf2a0447c0ac9f9e.js, which is a JavaScript file that runs within the browser. This makes it a prime candidate for caching, as it is inefficient to make a network request every time the page loads, even though the response stays identical. By caching this file, we skip additional HTTP requests when we need to load this JavaScript file, as it is readily available in our cache. Where is the cache located? In this case, the browser is using the disk as a cache. Even though it is slower than memory (RAM), it is still much faster than a network request.

## Diving deeper into caching

When a browser needs to load a resource, such as an image, it follows a sequence of steps to determine where to get the file:

- 1. Check the Memory Cache (It is using RAM like a cache, not that the RAM has a cache): The browser first checks its memory cache. This is used for resources downloaded in the current browsing session (since memory is non-persistent)
- 2. Check the Disk Cache: If the resource isn't in the memory cache, the browser checks the disk cache, a more persistent cache that contains resources from sites visited in the past
- 3. Network Request: If the resource isn't in either the memory or disk cache, the browser then makes a network request to the server hosting the resource.

When a cached file is found, it is known as a cache hit. When it isn't found, it is known as a cache miss.

## Cache Ratio

The cache refers to how many content requests a cache can fulfill successfully, relative to the number of request it recieves

This can be calculated by:

$$
\frac{\text{Cache Hit Rate}}{\text{Cache Hit Rate} + \text{Number of Cache Misses}}
$$

To convert this into percentage, it just times 100. The higher the percentage, the more preferable it is. This ratio is particularly important for CDNs (Content Delivery Networks).

## The server's perspective 

The server's perspective is more complicated than the client. When a client issues a 'getTweet()' request, how should the server determine which tweets to cache and which to retrieve from the disk?

- Write-around Cache: When a new tweet is posted, it will be written directly to the main storage (disk) rather than the cache. This then allows the cache space for more popular content that is accessed more frequently. However, if all of a sudden a tweet becomes popular and gains traction, it will be added to the cache after the first access. So when the tweet is accessed again, subseuqent reads will result in a cache hit. It will be a cache miss the first time and then onwards it will be found from the Memory, because we only want to store it in the memory only after it has been read.

- Write-through Cache: This will write to both the memory and the disk, regardless of whether the tweet is accessed or not. With this strategy, both the disk and the cache memory will have the same version of the tweet. However, this might put more load on the memory bus, filling the cache up with data that might not be accessed again.

- Write-back Cache: This will write data only to the cache initially. The data is written into the cache every time a change occurs but to the disk only when the cache is full.

## Eviction Polivies

An evicition policy is a system that determines which items get removed from the cache when the cache is full and new items need to be added.

- FIFO (First in First Out): This policy is similar to the queue interface. When the cahce becomes full, the first piece of data to be cached is removed first.

- LRU (Least Recently Used): The principle behind this method is that if an item that has not been accessed for a long time, it is less likely to be accessed in the future as well. Therefore the item should be removed from the cache. LRU caching woul d be useful if there were a single person with a really popular tweet, as we wouldn't want that tweet to be removed from our cache.

- LFU (Least Frequently Used): This evicts the items that are least frequently used. It assumes that if an item is not accessed frequently, it is unlikely to be accessed frequently in the future as well. LFU can be implmeneted using key-value pairs, where the key represents the item and the value represents the frequency of its usage. When the cache space runs out, the item wil the smallest frequest will be removed. While this approach seems reasonable, it has it's limitations. For example, tweets from 2013 that have a large number of view might never be evicted due to their frequency, which would prevent newer tweets from being stored.
