# HTTP

## Client-Server Model

**Client:** The client is an application or system that accesses services made available by a server. It can be a web browser, email software, or any application that needs to access a service. The client initiates communication by sending requests to the server for specific information or services.

**Server:** The server is a computer, device, or software that provides resources, data, or functionality to clients (or other servers) over a network. It listens for incoming requests and responds by processing those requests and returning a response.

**RPC (Remote Procedure Call)**

RPC allows a program to execute a function on a remote machine as if it were a local function call. For example, when you search "NeetCode" on YouTube, the browser displays a list of videos. The logic for retrieving these videos does not reside in the browser, but on YouTube’s servers. Behind the scenes, the client makes a request to the server (e.g. a function call like `listVideos("neetcode")`). The server processes this request and returns the results to the client. Although it appears as if the function is executed locally, it is actually executed remotely on the server, and the client receives the response over the network.

**How to identify an RPC:**
- It involves network communication (internet or local network)
- It depends on an external service (API, server, database)
- It may fail due to network issues (timeouts, connection errors)
- It is usually slower than local functions (due to network latency)

Examples:
- `requests.get(...)` → RPC (calls a remote server)
- `openai.chat.completions.create(...)` → RPC (remote API call)
- `cursor.execute(...)` → RPC if the database is remote

**Not RPC:**
- Pure Python functions (e.g. `add(a, b)`)
- In-memory operations (lists, dictionaries)
- File operations on your local machine

**Key rule:**
If the function requires sending data over a network to another machine to execute → it is RPC. Otherwise → it is local.

# HTTP (Hypertext Transfer Protocol)

HTTP is built on top of IP and TCP. At a basic level, IP (Internet Protocol) is responsible for delivering packets of data from the source to the destination, but it does not care about the order in which they arrive or whether they arrive at all. TCP (Transmission Control Protocol), which is built on top of IP, ensures that packets are delivered reliably and in the correct order. HTTP, which sits on top of TCP, is a request/response protocol. It defines the rules for how data is formatted and transmitted over the web.

## Anatomy of an HTTP Request
- **Request Method:** Indicates the desired action to be performed, such as GET, POST, PUT, or DELETE.
- **Request URL / URI:** Specifies where the request should be made (the resource being accessed).
- **Headers:** Provide additional information about the request or response.  
  - **HTTP Request Headers:** Contain information about the request, such as the method, the scheme (HTTP/HTTPS), and the types of data the client can accept.  
  - **HTTP Response Headers:** Provide extra information about the response or the server.
- **Body:** Not every HTTP request contains a body. GET requests typically do not have a body because they are only retrieving data, while POST and PUT requests usually include a body to send data to the server.

## HTTP Methods

- **GET:** It is used to retrieve a resource from the server. It is defined as idempotent, meaning that making the same GET request multiple times should always produce the same result and should not change the state of the server. 

- **POST:** It is used to send data to the server to create a new resource. A POST request typically contains a request body (payload) that includes the data being sent. This data is - placed in the body rather than the URL mainly for structure and size reasons, not specifically for security, as it can still be intercepted without HTTPS.

- **PUT:** It is used to update or replace a resource with new data. An important characteristic of PUT is that it is idempotent, meaning that performing the same update multiple times with the same data will result in the same final state of the resource.

- **DELETE:** It is used to remove a specified resource. It is also idempotent, meaning that deleting the same resource multiple times will result in the same final state (the resource no longer exists), even though the first request may return 200 (OK) and subsequent ones may return 404 (Not Found).

## HTTP Status Codes

HTTP status codes are used by the server to inform the client about the result of their request.

Informational Responses (100–199)
Informational responses indicate that the request has been received and the process is continuing.
- **100 Continue:** Indicates that the initial part of the request has been received and the client should continue sending the rest of the request.
- **101 Switching Protocols:** Indicates that the server is switching to a different protocol as requested by the client.

Successful Responses (200–299)
Successful responses indicate that the request was successfully received, understood, and processed.
- **200 OK:** The request has succeeded.
- **201 Created:** The request has succeeded and a new resource has been created, typically after a POST or sometimes a PUT request.

Redirection Messages (300–399)
Redirection responses indicate that further action needs to be taken by the client to complete the request.
- **300 Multiple Choices:** The request has more than one possible response, and the client should choose one.
- **301 Moved Permanently:** The requested resource has been permanently moved to a new URL.

Client Error Responses (400–499)
Client error responses indicate that there was an issue with the request made by the client.
- **400 Bad Request:** The server cannot process the request due to invalid syntax or malformed data.
- **401 Unauthorized:** Authentication is required or has failed for the requested resource.

Server Error Responses (500–599)
Server error responses indicate that the server failed to fulfill a valid request. These errors occur when something goes wrong on the server side while processing the client’s request.

# SSL/TLS

TLS (Transport Layer Security) ensures secure communication over a network. HTTP by itself is not secure and is vulnerable to man-in-the-middle (MITM) attacks. SSL (Secure Sockets Layer) is the older version of TLS and is now deprecated. Modern systems use TLS instead of SSL, although the term “SSL” is still commonly used. TLS provides encryption, which ensures that data transmitted between the client and server cannot be read by unauthorized parties. It also provides integrity, meaning the data cannot be altered during transmission, and authentication, which verifies the identity of the server. HTTPS is the combination of HTTP and TLS. It means that HTTP communication is encrypted and secured using TLS.

# API (Application Programming Interface)

An API (Application Programming Interface) provides a way for clients to interact with servers over a network. It defines a set of rules and protocols for building and interacting with software applications.

# The Three Paradigms

APIs can be classified into different paradigms such as REST, GraphQL, and gRPC. Each paradigm has its own characteristics and is suited for different scenarios. There are also other types such as SOAP (Simple Object Access Protocol) and Webhooks, but the 3 mentioned previously are the most commonly used.

## REST (Representational State Transfer)

REST refers to an API that follows the design principles and standards of the REST architectural style. It uses standard HTTP methods for communication between a client and a server.

### Advantages

REST APIs follow a client-server architecture where the client and server are separate entities communicating over a network. This separation allows independent development and scaling of both sides.

REST APIs are stateless, meaning each request from the client must contain all the necessary information for the server to process it. The server does not store information about previous requests. This makes REST APIs easier to scale horizontally because there is no need to manage session state on the server. This allows the loadbalancer to route the request to any server as well.

A common example of statelessness is pagination. For example, when retrieving videos from a server, the client may send requests such as:
- `https://youtube.com/videos?offset=0&limit=10`
- `https://youtube.com/videos?offset=10&limit=10`

In this case, the server does not remember previous requests. Instead, the client keeps track of the offset and sends it with each request.

### Disadvantages

One limitation of REST APIs is the issue of over-fetching and under-fetching data, especially in more complex applications.

Over-fetching occurs when the client receives more data than necessary. For example, if a client only needs a user’s profile picture, username, and comment, but the `/user` endpoint also returns additional fields like country or date joined, extra data is transferred unnecessarily.

Under-fetching occurs when the API does not provide enough data in a single request. This forces the client to make multiple requests to different endpoints to gather all required information, increasing latency and complexity.

## GraphQL APIs

GraphQL emerged as an alternative API paradigm to address the limitations of REST APIs, particularly the issues of over-fetching and under-fetching. With GraphQL, clients have the ability to precisely specify the required data within a single request, regardless of how the data is structured on the server. Following the client's specification, the server takes on the responsibility of gathering all the necessary data and formatting it accordingly.

In GraphQL, there are two main operations:
- Queries: These are used to retrieve data (used for GET)
- Mutations: These are used to modify data on the server (used for POST PUT, DELETE)

GraphQL operates through a single endpoint, typically an HTTP POST endpoint where all queries are sent (although GET can also be used in some cases). This API paradigm allows us to precisely specify the required data, reducing over-fetching and under-fetching. This approach can lead to more efficient data retrieval and reduced data usage.

## gRPC (Google Remote Procedure Call)

This is a framework for executing RPCs, a method that allows a program to execute a procedure in another address space. gRPC uses HTTP/2 over TCP, and HTTP/2 enables bi-directional streaming. It is typically used for server-to-server communication and is generally faster than REST APIs since it sends data using protocol buffers (these are language-neutral, platform-neutral, extensible mechanisms for serializing structured data), instead of JSON. gRPC also provides streaming, meaning we can send data from the client to the server and also from the server to the client. This is not a replacement for WebSockets, as WebSockets is a different protocol, while gRPC is built on top of HTTP/2.

# API Design

An API typically provides general CRUD (Create, Read, Update, Delete) operations. In the case of Twitter, these functions may include createTweet(), getTweets(), editTweet(), and deleteTweet().

These operations are performed on entities, which can be thought of as resources (tweets, users, etc.). The operations themselves, such as create and delete...., are equivalent to different REST methods such as GET, POST, PUT, and DELETE. As developers, we do not need to be concerned about the internal implementation of the API, but rather focus on the expected inputs and the return value.

## An example of creating a tweet

Creating a tweet uses the `createTweet(userId, content)` method. Making changes to an API method signature can break existing applications that depend on it. If we want to include a `parentId` parameter to become `createTweet(userId, content, parentId)`, it may disrupt a large number of users. Therefore, making the `parentId` parameter optional acknowledges that not every tweet is a reply and may not have a parent tweet. Most importantly, this ensures backward compatibility, which avoids breaking existing functionality for applications that rely on the original method.

When it comes to listing tweets, a different approach is needed. The idea of pagination is used, where parameters such as offset and limit are included:
https://api.twitter.com/v1.0/users/:id/tweets?limit=10&offset=0 This allows the REST API to remain stateless and prevents the client from retrieving too much data at once. By implementing pagination, it enables the retrieval of a specific number of tweets at a time while maintaining consistency in the returned data.

# API Versions

When significant changes are made to an API, such as adding a new parameter, method, or completely changing how things work, companies often update their API version. They may also announce that the previous version is deprecated, and developers are encouraged to update their API calls accordingly.

By versioning the API, it becomes easier to distinguish between different versions and helps developers adapt their code to accommodate any changes or improvements. Deprecating older versions prompts developers to transition to the latest version, which often offers better functionality, improved performance, and bug fixes.

The deprecation announcement serves as a notification for developers to update their code and align it with the latest API specifications. This ensures compatibility and a smooth transition while embracing advancements and updates in API design.

# CDNS

Content Delivery Network (CDNs), these are a group of cache servers that are located around the world so that they can cache content close to end users. Using a CDN, data can be accessed faster than an origin server. If a website is accessed throughout the world, it is beneficial to use a CDN, so even the clients that are far from the origin server can access the content without any delays. The kind of content that we would use CDN for would be static content such as HTML, CSS and static Javascript Files

In the diagram aboce, there are 7 CDN servers but in reality we could have 100s of them all around the world. The downside is that we can usually only put static content on them and we can't have application code on the CDN servers. This would mean that any computation or dynamic processing must still be handled by the origin servers. If a company that has multiple CDN servers, in the event that one of the CDN servers goes down, we can redirect to the next closer CDN server. Therefore by having multiple copie of the content distributed throughout the world, the system can withstand many more hardware failures.

## Different methods of using CDN (defining how content is being loaded into the CDN, choose only one)

### Push CDN

A push CDN is useful for website where the content is static and doesn't change frequeently. In a push CDN, once a new data is added to the origin server, it is immediately pushed to all of the cache servers. It is important to note that the content needs to be widely requested for a push CDN to be efficient. Since CDNs distribute content to all servers, they can be inefficient if the content isn't request by user near thoser servers.

A good example for a 'Push CDN' is websites that host video content consumed around the globe. Push CDNs offer an advantage in that users don't have to wait for the content to be transffered from the original server to the CDN. Each user request will result in a cache hit sinc ethe data has already been copied over.

## Pull CDN

If the data that a client or user is trying to access that does not exist on the CDN, a process similar to Write-around Cache. The CDN cache server will be checked first, and if the data is not found, then the cache server will retrieve it from the origin server and cache it for future requests.

In contrast to a push CDN, a pull CDN operates differently by requiring the CDN to pull the content from the origin server on an "as needed" basis. 

Websites like Twitter are ideal candidates for pull CDN due to the large amount of content genreated every second. It will be a significant burdnen for the Twitter team to proactively opush all the content to each CDN. Instead. when a user requests content from a server, the CDN will automatically retrieve it from the server and store it in the cache. Furthermore, different users in different regions of the world may request different content. As a result, different CDNs located in different regions may have distinct content tailored to their respective audiences. (This is typically more relevant to larger platforms with geo-specified content)
