# Computer Architecture

**Disk (persistent storage)** Two different types:  
* **HDD (Hard Disk Drive):** Mechanical and uses spinning disks. It is generally slower due to mechanical movement, and performance can degrade over time due to wear and tear.  
* **SSD (Solid State Drive):** Uses flash memory (no moving parts) and reads/writes data electronically, making it much faster and more durable.

**RAM (Random Access Memory):** Reading and writing to RAM is faster than doing it to disk. It stores the applications currently running and the variables your program uses. The data is erased once the computer is shut down. They typically come in sizes from 1GB to 128GB or more.

**CPU (Central Processing Unit):** The brain of the computer that reads and executes instructions. It performs computations such as addition, subtraction, multiplication, and logical operations, and works with data stored in RAM (and cache).

**Cache:** Cache uses SRAM and is located inside the CPU (L1, L2, L3 cache). It is faster than RAM but much smaller (typically KBs to tens of MBs). When a read operation is requested, the system checks the cache first, then RAM, then disk. If the data is found in the cache and is valid, it is fetched from the cache instead of RAM.

# Design Requirements

A good metric used to evaluate if a system design is good is **availability**.

It can be calculated as:
$$
\text{availability} = \frac{\text{uptime}}{\text{uptime} + \text{downtime}}
$$

A common target for companies is **99.999% availability ("five nines")**, which corresponds to about **5 minutes of downtime per year**. Availability is used to define SLOs (Service Level Objectives) and SLAs (Service Level Agreements).

**SLA (Service Level Agreement):** A formal, external agreement between a company and its customers that defines guaranteed service levels (e.g. uptime, response time). It usually includes consequences such as penalties or compensation if the guarantees are not met.

**SLO (Service Level Objective):** An internal target that a team aims to achieve in order to meet the SLA. It defines specific measurable goals (e.g. 99.9% uptime, 200ms response time) and is used for monitoring and engineering decisions. SLOs are typically set stricter than SLAs to provide a safety margin.

**Throughput:** The amount of work a system can handle over a period of time (e.g. requests per second).

**Latency:** The time it takes for a single request to be processed from start to finish (often measured as response time).

**Vertical Scaling:** Increasing the resources (e.g. CPU, RAM) of a single server to handle more load.

**Horizontal Scaling:** Increasing capacity by adding more servers to distribute the load.
