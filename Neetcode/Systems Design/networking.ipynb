# Networking Basics

An IP address is a distinct numeric identifier for every device connected to a computer network.

**IPv4:** A 32-bit address expressed in the range of 0.0.0.0 – 255.255.255.255, consisting of four octets where each octet is composed of 8 bits.  

**IPv6:** A 128-bit address consisting of 8 groups of 16 bits, written in the format (xxxx:xxxx:xxxx:xxxx:xxxx:xxxx:xxxx:xxxx).

The IP address operates at the **network layer** of the OSI (Open Systems Interconnection) model.

![alt text](image/image2.png)

**Public IP Address:** A public IP address is a globally unique identifier assigned to a network by an ISP (Internet Service Provider). It is reachable from any device on the internet.  
Example: `8.8.8.8`

**Private IP Address:** A private IP address is used within a LAN (local area network) and is not directly reachable from the internet. The same private IP ranges can be reused across different networks.  
Example: `192.168.1.10`


**Static IP Address:** An IP address that does not change over time. It is manually assigned or permanently reserved (usually by an ISP (Internet Service Provider) or network administrator). It is commonly used for hosting services, servers, or remote access.

**Dynamic IP Address:** An IP address that can change over time. It is automatically assigned using from a pool of available addresses. Most home users have dynamic IPs.

**Ports:** Network ports are numeric identifiers used to distinguish different applications or services running on a device and to identify specific connections in network communication. A port is associated with a service (a program listening on that port), which can handle many requests and return different content.


```
Request 1 → google.com → returns Google page
Request 2 → amazon.com → returns Amazon page
```


Example:

Inside a home network:
```
Laptop   → 192.168.1.10
Phone    → 192.168.1.11
Router   → 192.168.1.1
```

From the internet, all devices appear as:
Each connection from devices inside the network is mapped to a unique port on the public IP. These ports are used by the router (via NAT) to distinguish between different connections.

```
Public IP → 203.x.x.x
```

The router uses NAT (Network Address Translation) to map:

```
192.168.1.10:5000 → 203.x.x.x:60001
192.168.1.11:5000 → 203.x.x.x:60002
```

This allows multiple devices with private IPs to share a single public IP.

**TCP (Transmission Control Protocol)**

TCP ensures reliable delivery of data by retransmitting packets that are lost in the network. Lost packets are those that were sent by the sender but never received by the receiver.

It establishes a connection using a **3-way handshake**:

1. The client sends a SYN packet to the server to initiate a connection.  
2. The server responds with a SYN-ACK, indicating it received the request and is ready to communicate.  
3. The client sends an ACK to confirm receipt of the server’s response.

This is more reliable than a 2-way handshake because it ensures both sides confirm that communication is established. In a 2-way handshake, the server may respond to the client, but if that response is lost in the network, the client never receives it. As a result, the server assumes the connection is established while the client does not, creating a half-open connection.

Although TCP provides relaibility, it does have its drawbacks and brings about more overhead and tends to be slower due to its rigorous apporach to ensure delivery

**UDP (User Datagram Protocol)**

UDP is a connectionless transport protocol that prioritizes speed over reliability. It sends data without establishing a connection beforehand and does not guarantee delivery, ordering, or duplication protection.

Unlike TCP, UDP does not retransmit lost packets or reorder them. Each packet is sent independently, and the receiver processes packets as they arrive.

UDP is commonly used in scenarios such as online gaming and video streaming, where low latency is more important than perfect reliability. In these cases, it is often preferable to skip a lost packet (e.g. a dropped frame) rather than delay transmission by retransmitting it.

UDP works as follows:
1. The sender packages data into a datagram with a destination IP address and port.
2. The datagram is sent directly over the network without establishing a connection.
3. The network delivers the packet on a best-effort basis (it may be lost, duplicated, or arrive out of order).
4. The receiver accepts the packet and passes it to the application without acknowledgment.

Therefore, UDP is chosen in situations where speed and efficiency take precedence over reliability and error correction.

**DNS (Domain Name System)**

DNS is like the internet’s phone book. It operates as a naming system that converts easily readable domain names into numerical IP addresses. When we input a domain name such as `google.com` into a web browser, DNS translates it into an IP address (e.g. `142.251.211.238`).

DNS does not determine the port; it only resolves the domain name to an IP address. The port is determined by the protocol scheme used, such as HTTP (port 80) or HTTPS (port 443), or can be explicitly specified by the user.

The overall coordination and management of domain names within the DNS system is overseen by ICANN (Internet Corporation for Assigned Names and Numbers).

ICANN oversees the global domain name system, similar to how a shopping mall management oversees the entire mall. When registering a domain, users interact with domain registrars (e.g. GoDaddy, Google Domains), which are authorized by ICANN. These registrars handle domain name searches, registration, and record management. ICANN does not sell domain names directly; it coordinates and regulates the system, while registrars provide the actual domain registration services.

(TLDR) \
Domain ↔ IP is many-to-many, not one-to-one


**Registry vs Registrar vs DNS**

**Registry:** The registry is the authoritative organization that manages a TLD (Top-Level Domain) such as `.com`, `.org`, or `.sg`. It maintains the official database of all registered domain names under that TLD.  
Example: Verisign manages `.com`

**Registrar:** A registrar (e.g. GoDaddy, Google Domains) is a company accredited by ICANN that allows users to search, purchase, and manage domain names. The registrar communicates with the registry to register the domain on your behalf.

**DNS (Domain Name System):** DNS is the system that translates domain names into IP addresses. After a domain is registered, DNS records (e.g. A record) are configured to point the domain to a specific server.

---

**How they work together:**

1. You buy a domain from a registrar (e.g. GoDaddy)  
2. The registrar registers it with the registry (e.g. .com registry)  
3. You configure DNS to point the domain to an IP address  
4. When users visit the domain, DNS resolves it to the server’s IP  

---

**Summary:**

- Registry → stores official domain records  
- Registrar → sells and manages domains  
- DNS → translates domain → IP address

# Anatomy of a URL (Uniform Resource Locator)

Given: `https://domains.google.com/get-started`

![alt text](image/url.png)

**Protocol Scheme:** URLs commonly start with a protocol such as HTTP or HTTPS, which indicates how the resource should be accessed.  
- HTTP → port 80  
- HTTPS → port 443  

This specifies which protocol the browser should use to communicate with the server.

Other protocols include:

**FTP (File Transfer Protocol):** Used to access and transfer files between systems.

Example:
ftp://ftp.example.com/pub/file.txt

- `ftp://` → protocol  
- `ftp.example.com` → FTP server domain  
- `/pub/file.txt` → path to the file on the server  

**SSH (Secure Shell):** Used to establish secure remote connections to servers.

Example:
ssh://username@example.com:22

- `ssh://` → protocol  
- `username` → user for authentication  
- `example.com` → server domain  
- `:22` → port number (default for SSH)  

**Domain Name**

A domain name can be divided into 3 components: the subdomain, the primary domain, and the top-level domain (TLD).

**Subdomain:** A subdomain is a prefix added to the main domain that allows you to organize or separate different sections of a website under the same domain. It represents a separate logical section but is still part of the same overall domain. They are used to separate different services or parts of a system under the same main domain (e.g. `api.example.com`, `mail.google.com`). And are typically used when components need to be logically or technically independent, such as different applications, environments, or infrastructure.

Example:
```
blog.example.com
api.example.com
```

Here:
- `blog` and `api` are subdomains
- `example` is the primary domain
- `.com` is the TLD

A subdomain is different from a subdirectory (e.g. `example.com/blog`), which is just a path within the same website rather than a separate section.

**Primary Domain:** The primary domain (also called the second-level domain) represents the core identity of the website. When you purchase a domain, you own this part and can create multiple subdomains under it.

**Top-Level Domain (TLD):** The `.com` portion is the TLD, which is the highest level in the domain hierarchy. TLDs help categorize domains (e.g. `.com`, `.org`, `.sg`), but they do not strictly enforce the type of website.

**Paths:** Paths are part of a URL and are denoted using forward slashes `/`. They represent a specific location or route within a website where particular content or resources are located. 

Example:  
`/get-started` indicates a specific section or resource within the website. By including a path in the URL, users can directly access that content.

**Ports:** A port number can be specified in a URL, but it is usually omitted because standard protocols use default ports:
- HTTP → port 80  
- HTTPS → port 443  

If a non-standard port is used (e.g. 8080), it must be explicitly included in the URL.

Example:  
`http://localhost:8080`

This means the server on `localhost` is listening for HTTP requests on port 8080 instead of the default port 80. `127.0.0.1` is neither a public IP nor a typical private IP. It is the **localhost IP address**. It is a **loopback address**, which is used to refer to your own computer. Loopback addresses are reserved for internal communication within the same device and are not reachable from any network (including your local network or the internet).
