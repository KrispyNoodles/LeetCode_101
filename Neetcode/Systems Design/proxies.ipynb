# Proxies

## Forward Proxies

Forward proxies is when your computer (the sender) wants to access a website (the recepient), it sends the request to the proxy server (your friend). The proxy server then sends the request to the website on your behalf. The proxy server masks your IP addresses and uses its own when communicating with the website, keeping your address hidden. When requesting frequently for the same data, the proxy server can cache the response and provide it to you without having to request the same data from the webstie again. A proxy server can also filter your request such as restricting access to certain websites or resources, similarly to how a company or school might block access to entertainment sites.

A forward Proxy serves as an intermediary in network communication:
- 1. Providing privacy
- 2. Increasing efficency through caching
- 3. ALlowing control over network traffic

![alt text](image/fw_proxy.png)

A VPN is similar to a forward proxy in that it hides the client’s IP address, but it operates at the network level and routes all traffic, not just HTTP requests (only forward proxy).

## Reverse Proxy

A reverse proxy anaoymizes the destination server instead of the client. It also known as an inbound proxy. Instead of recieving requests from clients and forwarding them, it receives requests and forwards to the appropriate server. The benefit of a reverse proxy is its ability to protect servers by managing incoming request and distributing the load across multiple servers. It provides protection against Distributed Denial of Service (DDoS) attacks, because the Proxy Server know which IP's are requesting how many times, it can restrict that IP address by allowing a maximum of 100 request per IP. This acts as an additional layer of secutiy and privacy for the servers. 

An example is CDN (Content Delivery Networks), instead of clients directly acessing the origin server, the reverse proxy (CDN) handles the requests. If the request content is available, the reverse proxy servers it directly to the client, effectively reducing latency and alleivating the load on the origin server.

![alt text](image/rv_proxy.png)

TLDR: A forward proxy sits on the client side and hides the client’s identity, while a reverse proxy sits on the server side and hides backend servers and manages incoming requests.

# Load Balancers

An example of using the reverse proxy is a load balancer. It resides between the client and the server, functioning as an intermediary for their communication. Load balancing can be implmeneted using different strategies to efficiently distribute incoming network requests. They are designed to optimize resource use, prevent overloading and improve the overall system's responsiveness and performance


## Round Robin (RR)

Round robin is a technique commonly used for distributing requests across multiple server in a sequential manner.

![alt text](image/rr.png)

## Weighter Round Robin (WRR)

THis strategy allows the network traffic to be distributed based on the computational power of each server. Computational power entails of the server's CPU processing Power, RAM, storage and network bandwidth. This technique dsitrubtes requests proportionally, and ensures a more efficient use of resources.

![alt text](image/wrr.png)

## Least Number of Connections

This strategy is useful when there are significant differences or unpredeictable variations in the time required to process each request. This works by assigning new requests to the server with the fewest active connections.

An example is during peak shopping seasons like Black Friday or Cyber Monday, e-commerce websites often experience high traffic volumes. Some users may simply browse products, while often may perform payment transations, add multiple items to their cart or apply various filtersw to their search. In his case, let's assume that Server A is handling complex checkout requests and Server C is responsible for handling subscriptions. The load balancer may assign a new request to server C to ensure that Server A or B is not overaloaded.

By dynamically distributing requests to servers with the fewest active connections, the least connections method hels prevent imbalances and optimizes resource utilization within a load-balanced environment.

## User Location

For this strategy, we want to ensure that users recieve data from the nearest server, minimizing the distance that data needs to travel. This approach helps reduce latency and provides a faster and smoother user experience

# Layer 4 and Layer 7 Balancing

## Layer 4 Load Balancing
In the OSI (Open Systems Interconnection) model, the network layer model specifically, layer 4 corresponds to the transport layer which is responsible for end-to-end communication. This layer manages protocols like TCP and UDP. Layer 4 load balancers are designed to route traffic based on layer 4 data, such as IP address and TCP port, rather than the contents of the requests. This makes Layer balancing fast and straightfoward.


## Layer 7 Load Balancing
Layer 7 balancers operate at the application layer of the OSI model. This layer is responsible for providing services to the software application that initiates the network request. Layer 7 load balancers analze and route traffic based on data present in the content of the requests, focusing on HTTP headers, methods or body content. This introduces extra processing overhead, but it offers greater flexibility and sophisitication in routing decisions. This can enable advanced handling of requests based on speicfic content charctersistics.


# Consistent Hashing

Another technique that can be used to map requeswts to servers in the context of load balancing is hashing. Each request will have an IP address, user ID or a request ID associated with it. If we can calculate the hash of this unique identifier and mod it by the number of servers we have available, the answer will allow us to assign the request to a specific server.

For example, if we have servers numbered 0,1,2. And we take the unique identifier of three incoming requests, being 6,7,8. 

The mapping becomes:
```python
6 % 3 = 0
7 % 3 = 1
8 % 3 = 2
```

You would notice that the same IP addres will always be redirected to the same server. The benefit of this is that then each server can cache data that belongs to a specific user. This makes sense because it is expected that the user 6 will always be direct to server 0. This is different from the round robin in the sense that it is consistent in assigning the user with the same IP address to the same server every time.

## Issue with regular hashing

However the issues comes if for example server 2, goes down. This arithmetic has to be done again, resulting in a new map. IP address 9 and 11 will be mapped to 0 and IP address 10 will be mapped to 1. But this becomes an issues as the cache data that was stored previously is mismatched to the current IP addresses assigned to each servers.

# Consistet Hashing

This uses a ring-based structure and a hash function to map requests to servers. If a server goes down, the requests it handled are reassigned to the next available server in a clockwise direction on the ring.

The ring represents the space of all possible hash values. Each server in the system is also represented as a point on this ring. The hash function takes the content of the request (IP address) and maps it to a hash value within the range of the ring. If a server goes down, its portion of the ring becomes unavailable. 

In consistent hashing, the requests that were previously handled by that server are reassigned to the next closest server in a clockwise direction on the ring.

Consistent hashing can also be applied in CDNs (Content Delivery Network), and even databases where a user's data resides on a specifc server, and consistent routing of that user to that server is desired.
