```python
# Bucket Sort
# we usually use a bucket sort when we know the certain range of values that are in the array that we are tryign to sotr

array_1 = [2,1,2,0,0,2]

# for each value in the array we are counting how many times they do appear

bucket_arr = [0, 0, 0]
```

```python
for value in array_1:
    bucket_arr[value] += 1

print(bucket_arr)

index = 0

# unique values
for i in range(len(bucket_arr)):

    for _ in range(bucket_arr[i]):
        array_1[index] = i
        index +=1

print(array_1)

# complecity of O(n) since it loops through once to count and another time to put the valeus back in
# unstable sorting algo as we dont care where were the previous arrangment of the previous values were at previously
```
