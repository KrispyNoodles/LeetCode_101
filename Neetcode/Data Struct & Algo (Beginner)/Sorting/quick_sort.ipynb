```python
# QUICK SORT

# it is similar to the merge_sort, but instead of splitting into half always
# what it does is to use the right value of the array as the pivot and do a comparison as too whether it is larger or smaller than the pivot

# Quick Sort
def quickSort(arr, s, e):

    # this is to check for cases where it is just a single elemtn
    # it is the same means it is the same element
    # and also catches the segments where it is empty
    if s>=e:
        return arr
        
    # the pivot value is the value at the end of the array
    pivot = arr[e]
    left = s

    for i in range(s, e):
        
        if pivot > arr[i]:

            # do the swap
            arr[i], arr[left] = arr[left], arr[i] 

            # increment s
            left += 1
    
    # after evertyhing swap the last i with the pivot value
    arr[left], arr[e] = arr[e], arr[left]


    # running quicksort on left and right after the pivot has been placed
    quickSort(arr, s, left-1)
    quickSort(arr, left+1, e)

    return arr


arr = [1,3,4,5,7,2,1]
s = 0
e = len(arr)-1

quickSort(arr, s, e)
```
