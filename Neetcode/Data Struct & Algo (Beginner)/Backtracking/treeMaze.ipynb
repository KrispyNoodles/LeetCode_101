# Tree Maze

```python
# creating a tree
# creating a TreeNode
class TreeNode:

    # its own value, left and right nodes
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None

# building a tree
t4 = TreeNode(4)

t0 = TreeNode(0)
t1 = TreeNode(1)

t7 = TreeNode(7)
t2 = TreeNode(2)
t0_2 = TreeNode(0)

t4.left = t0
t4.right= t1

t0.right = t7

t1.left = t2
t1.right = t0_2
```

```python
# trying to create a function to check if a path exists from the root of the tree to the leaf node, won't go pass any zeroes

def canReachLeaf(root):

    if not root or root.val==0:
        return False

    # it has reacheded the bottom
    if not root.left and not root.right:
        return True

    # it moved downwards the left part with a True
    if canReachLeaf(root.left):
        return True

    if canReachLeaf(root.right):
        return True

    return False

canReachLeaf(t4)
```

```python
# variation with the values of the path it moves across
path = []

def canReachLeaf(root):

    if not root or root.val==0:
        return False
    
    path.append(root.val)

    # it has reacheded the bottom
    if not root.left and not root.right:
        return True

    # it moved downwards the left part with a True
    if canReachLeaf(root.left):
        return True

    if canReachLeaf(root.right):
        return True

    # if it fails, pop it to move back up once
    path.pop()
    return False

canReachLeaf(t4)
print(path)
```
