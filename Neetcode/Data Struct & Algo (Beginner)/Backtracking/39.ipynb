# 39. Combination Sum

```python
from typing import List

class Solution:
    def combinationSum(self, nums: List[int], target: int) -> List[List[int]]:
        
        path = []
        master_set = []

        def helperFn(start, target):

            if target==0:
                master_set.append(path.copy())

            # when it reaches negative, it means that it is minusing too much
            # it should stop exploring and go back
            if target<0:
                return
            
            for i in range(start, len(nums)):

                # select an option
                path.append(nums[i])

                # explore other options, with the updated target
                helperFn(i, target-nums[i])

                path.pop()
                

        helperFn(0, target)

        return master_set
```

```python
s = Solution()

# Input:
nums = [2,5,6,9]
target = 9

output = s.combinationSum(nums, target)
output
```
