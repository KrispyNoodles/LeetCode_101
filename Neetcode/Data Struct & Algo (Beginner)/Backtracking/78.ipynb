```python
array_1 = [[],[1,2],[1,3]]

add_array = [1,3]

# it does not add [1,3] but adds [1,3,2]
# because it is appending a reference rather than a copy
# array_1.append(add_array) ## WRONG
array_1.append(add_array.copy()) ## CORRECT 

add_array.append(2)

array_1.append(add_array)
array_1
```

```python
from typing import List

class Solution:
    def subsets(self, nums: List[int]) -> List[List[int]]:

        master_set = []
        temp_array = []
        
        def helperFn(start):
            
            # has to be a copy to add in array at that point
            # if not it is appending a reference
            master_set.append(temp_array.copy())

            for i in range(start, len(nums)):

                temp_array.append(nums[i]) # select option
                helperFn(i+1) # explore
                temp_array.pop() # backtrack, it undos your option when coming back

        helperFn(0)

        return master_set

# time complexity of O(n*2^n)
```
