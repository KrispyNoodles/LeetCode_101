## HashMap Basics

A hashmap stores data as key-value pairs.

When inserting a key:
1. The key is converted into an integer using a hash function.
2. That integer is used to determine the index in an underlying array.


Example
```python
hashMap = {}
hashMap["Alice"] = "NYC"
hashMap["Brad"] = "Chicago"
hashMap["Collin"] = "Seattle"
```

Assume the hashmap starts with size 2 and "Alice" -> 25

25 % 2 = 1

Index | Key, Value
|-----|-----------|
0     |
1     | Alice, NYC

After insertion, the hashmap resizes (e.g., doubles to size 4). We must rehash existing keys:

25 % 4 = 1

Index | Key, Value
|-----|-----------|
0     |
1     | Alice, NYC
2     |
3     |


Adding Another Key, "Brad" -> 27

27 % 4 = 3

Index | Key, Value
|-----|-----------|
0     |
1     | Alice, NYC
2     |
3     | Brad, Chicago


### Collision Handling

When two keys map to the same index, a collision occurs.

Example:
If "Collin" hashes to index 1, it collides with "Alice".


### Collision Resolution Strategies

1. Chaining

- Each index stores a list (or linked list) of key-value pairs.
- Multiple items can exist at the same index.

Example:
Index 1:
(Alice, NYC) -> (Collin, Seattle)

- Lookup requires traversing the list.


2. Open Addressing

- Instead of storing multiple items in one slot, find another empty slot.
- Move to the next available index.

Process:
1. Compute index using hash.
2. If occupied, check next index.
3. Repeat until an empty slot is found.

````python
class Pair:
    def __init__(self, key, val):

        self.key = key
        self.val = val
    
class HashMap:
    def __init__(self):
        self.size = 0
        self.capcaity = 2
        self.map=[None, None]

    def hash(self, key):
        index = 0

        for c in key:
            # ord converts to ascii value
            index += ord(c)

        return index%self.capcaity

    # Open addressing concept
    def get(self, key):

        # find the index
        index = self.hash(key)

        while self.map[index] != None:

            # if it is not, then move down to the next index
            if self.map[index] == key:
                return self.map[index].val
            index+=1

            ## in case we move out of bound of the capacity
            index = index%self.capcaity
        
        # if no key exist
        return None

    def put(self, key, val):

        index = self.hash(key)

        while True:
            # if the key does not exist, means we can put it in
            if self.map[index] == None:
                self.map[index] = Pair(key, val)
                self.size+=1

                # rebuild the entire hash if the size is bigger than half of the capacity
                if self.size >= self.capcaity//2:
                    self.rehash()
            
                return
            
            # if there is a key, we will change the value that was attached to the key
            elif self.map[index].key == key:
                self.map[index].key = val
                return
        
            # else increment to the next index
            index+=1

            # incase it goes oob
            index = index%self.capcaity

    def rehash(self):
        self.capcaity = 2 *self.capcaity

        newMap = []

        for _ in range(self.capcaity):
            newMap.append(None)


        oldMap = self.Map
        self.Map = newMap

        # reset to 0 because the put function will increment it anwyay
        self.size = 0

        for pair in oldMap:
            if pair:
                self.put(pair.key, pair.val)

````
