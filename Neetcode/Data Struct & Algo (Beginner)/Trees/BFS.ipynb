# BFS (Breadth First Search)

Row by row, and left to right



```python
# creating a tree
# creating a TreeNode
class TreeNode:

    # its own value, left and right nodes
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None

# building a tree
t2 = TreeNode(2)
t3 = TreeNode(3)
t4 = TreeNode(4)
t5 = TreeNode(5)
t6 = TreeNode(6)
t7 = TreeNode(7)

t4.left = t3
t3.left= t2

t4.right = t6

t6.left = t5
t6.right = t7

 
```

```python
from collections import deque

def bfs(root):

    queue = deque()
    
    # if root exist add it to the queue
    if root:
        queue.append(root)

    level = 0 

    # as long as there is values in the queue, run
    while len(queue)>0:
        print(f"Level: {level}")

        for _ in range(len(queue)):

            # following FIFO
            curr = queue.popleft()
            print(curr.val)

            if curr.left:
                # append as there might be more laters
                queue.append(curr.left)

            # cant use elif as some nodes have both left and right
            if curr.right:
                queue.append(curr.right)
        
        level+=1

bfs(t4)
```
