# DFS (Depth-First-Search)

For every single leaf node we are going as deep as we can

All complexity is O(n)




inOrder: [2, 3, 4, 5, 6, 7]

revOrder: [7, 6, 5, 4, 3, 2]

preOrder: [4, 3, 2, 6, 5, 7]

postOrder: [2, 3, 5, 7, 6, 4]

```python
# creating a tree
# creating a TreeNode
class TreeNode:

    # its own value, left and right nodes
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None

# building a tree
t2 = TreeNode(2)
t3 = TreeNode(3)
t4 = TreeNode(4)
t5 = TreeNode(5)
t6 = TreeNode(6)
t7 = TreeNode(7)

t4.left = t3
t3.left= t2

t4.right = t6

t6.left = t5
t6.right = t7

 
```

```python
# printing out the values in the nodes after building the tree

# inorder

def inOrder(root):

    if not root:
        return
    
    inOrder(root.left)
    print(root.val)
    inOrder(root.right)

inOrder(t4)
```

```python
# reverse order

def revOrder(root):

    if not root:
        return
    
    revOrder(root.right)
    print(root.val)
    revOrder(root.left)

revOrder(t4)
```

```python
# preorder, going to the root then down the left, then going down the right
# root.val -> left.val -> right.val
# then move to left node

def preOrder(root):

    if not root:
        return
    
    print(root.val)
    preOrder(root.left)
    preOrder(root.right)

preOrder(t4)
```

```python
# postorder

def postOrder(root):

    if not root:
        return
    
    postOrder(root.left)
    postOrder(root.right)
    print(root.val)

postOrder(t4)
```
