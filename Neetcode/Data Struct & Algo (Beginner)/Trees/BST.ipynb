```python
# Binary Search Tree (BSTs)

# creating a TreeNode
class TreeNode:

    # its own value, left and right nodes
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None

# similarly to binary search
def search(root, target):

    # this means the tree has ended
    if not root:
        return False

    # this means the target is on the right side of the tree
    if target > root.val:
        return search(root.right, target)

    # target is on the left side
    elif target < root.val:
        return search(root.left, target)

    else:
        # meaning it is able to find the target in the tree
        return True

# time complexity is also O(log(n)) too, similarly to binary search

# extra info
# The root node is typically decided by the order of insertion—the very first value added to the empty tree automatically becomes the root
```

```python
# building a tree
t1 = TreeNode(1)
t2 = TreeNode(2)
t3 = TreeNode(3)
t4 = TreeNode(4)

t2.left = t1
t2.right= t3

t3.right = t4

# t2 is the root node, and 4 is the target value
answer = search(t2,4)
print(answer)

# t2 is the root node, and 5 is the target value
# 5 is not inside the tree so will return false
answer = search(t2,5)
print(answer)
```

### Why use Bin Search Tree instead of Arrays

When inserting/deleting an element from an Array is an O(n) operation, because all the other elements have to be shifted.
Whereas inserting/deleting an element from a BST is an log(n) operation.

```python
# adding into a tree
# it is easy as it just follows the value and the direction
# an assumption is that the value being added does not exist in the bft

def insert(root, val):

    # if the root is None, it means it has finally reached where the node should be create
    if not root:
        return TreeNode(val)
    
    # move down the right path
    if val>root.val:
        root.right = insert(root.right, val)

    # move down the left path
    elif val<root.val:
        root.left = insert(root.left, val)
    
    return root

# time complexity of O(h), where h is the height of the tree
# because at each recursive call, we move down one level along a single path from root to leaf
```

```python
# Removing a node from a tree is difficult
# the first step is to identify the largest value of the left side of the root or the smallest value on the right side of the root, 
# either these two values can be replaced with the root node to ensure the properties of the bst remains the same

# returning the minimum value of the BST
def minValueNode(root):
    cur = root

    while cur and cur.left:
        cur = cur.left

    return cur


# removing a node and returning the root of the BytesWarning
def remove(root, val):

    if not root:
        return None
    
    # these two if statements are used to identify which part of the tree the node to be deleted is at
    if val>root.val:
        root.right = remove(root.right, val)

    elif val<root.val:
        root.left = remove(root.left, val)

    # once it is found, then this will be activated
    else:
        # if there is no more on root left
        if not root.left:
            return root.right
        elif not root.right:
            return root.left
        else:
            # getting the smallest value on the right to become the new root of the entire BST
            minNode = minValueNode(root.right)
            root.val = minNode.val

            # removing the value, because it is now in the root and existent as a leaf below
            root.right = remove(root.right, minNode.val)

    return root

# time complexity of O(log(n)) is tree is balanced and O(n) if tree is unbalanced
# The time complexity of inserting and removing from a BST is proportional to the height of the tree.
```
