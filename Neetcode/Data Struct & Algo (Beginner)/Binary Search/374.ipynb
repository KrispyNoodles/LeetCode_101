# 374. Guess Number Higher or Lower

```python
# The guess API is already defined for you.
# @param num, your guess
# @return -1 if num is higher than the picked number
#          1 if num is lower than the picked number
#          otherwise return 0
# def guess(num: int) -> int:

# takes too slow and faced TLE
# n is the max number
class Solution:
    def guessNumber(self, n: int) -> int:

        answer = False

        while answer != True:
            
            # this means the guess is too high
            if self.guess(n) == -1:

                # decrease the guess number
                n -= 1

            elif self.guess(n) == 1:
                n += 1

            # else the guess is found
            else:
                return n
    
    def guess(self, num: int):
        
        secret_num = 10

        if num > secret_num:
            return -1

        elif num < secret_num:
            return 1

        else:
            return 0

# time complexity of O(n)
```
