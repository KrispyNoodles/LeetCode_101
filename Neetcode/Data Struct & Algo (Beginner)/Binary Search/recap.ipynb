```python
# assumign that the avalue is inside always
def bin_seach(array, target):

    left = 0
    right = len(array)-1

    while right>=left:

        # updating the middle
        middle = (right+left)//2

        if target == array[middle]:
            return middle
        
        elif target > array[middle]:
            # move the left
            left = middle+1

        else:
            right = middle-1
```

```python
# return the next biggest
def bin_seach(array, target):

    left = 0
    right = len(array)-1

    while right>=left:

        # updating the middle
        middle = (right+left)//2
        
        if target >= array[middle]:
            # move the left
            left = middle+1

        else:
            right = middle-1

    # answer is rarely from the middle, usually from either left or right
    # left is round to the next right, or place outside
    # right is round to the next left, or place outside
    return left
```
