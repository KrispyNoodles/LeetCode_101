```python
# now a range is given instead of a target??

# normal bin search
def binarySearch(low, high):

    while high>low:

        # calculation of mid
        middle = (high+low)//2

        # this means the value is too high
        if isCorrect(middle) > 0:
            high = middle - 1

        # this means the value is too low
        elif isCorrect(middle) < 0:
            low = middle + 1

        else:
            return middle
    
    # the value cant be found
    return -1

# simple helper function to check for n
def isCorrect(n):

    secret_taget = 10

    if n > secret_taget:
        return 1

    elif n < secret_taget:
        return -1
    
    # this mean it has been found
    else:
        return 0

binarySearch(1,100)
```
