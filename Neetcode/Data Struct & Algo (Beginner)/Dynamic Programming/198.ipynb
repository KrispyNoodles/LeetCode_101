# 198. House Robber

```python
# brute force
class Solution:
    def climbStairs(self, n: int) -> int:
        
        # if you are at stair 1, you have 1 method only
        if n==1:
            return 1
        
        # if you are at stair 2, you have 2 methods 1+1 and 2
        if n==2:
            return 2
        
        return self.climbStairs(n-1) + self.climbStairs(n-2) 

# time complexity O(2^n)
```

```python
# having a saved memory
from typing import List

# TLE issue
# this decides if robbing the current or the next is better
# but it is the max of either f(n)=max((n+f(n+2)), f(n+1))
class Solution:
    def rob(self, nums: List[int]) -> int:

        def helperFn(index):

            # checking index is within range, if not no money to rob
            if index>=len(nums):
                return 0
        
            # can choose to either rob current house or next house
            # rob current or skip current
            total_money = max(nums[index] + helperFn(index+2), helperFn(index+1))
            
            return total_money

        # starting with index 0
        return helperFn(0)


# time complexity of O(n^2)
# space complexity of O(n)
```

```python
from typing import List

# Memoization/ Top Down Dynamic Programming
class Solution:
    def rob(self, nums: List[int]) -> int:

        def helperFn(index, cache):

            # checking index is within range, if not no money to rob
            if index>=len(nums):
                return 0
            
            # checking if the answer exists in the cache
            if index in cache: 
                return cache[index]
        
            # can choose to either rob current house or next house
            # rob current or skip current
            cache[index] = max(nums[index] + helperFn(index+2, cache), helperFn(index+1, cache))
            
            return cache[index]

        # starting with index 0
        return helperFn(0, {})


# time complexity of O(n)
# space complexity of O(n)
```
