solving the fibonacci sequence

F(0)=0, F(1)=1,

F(n)=F(n-1)+F(n-2)

to solve F(5)

![alt text](fibo.png)

```python
# brute force method to solve

def bruteForce(n):

    # if the sequence is 0 return 0
    # if it is 1 return 1
    if n==0:
        return 0
    elif n==1:
        return 1
    
    # else return the function of F(n-1) and F(n-2)
    else:
        return bruteForce(n-1) + bruteForce(n-2)

bruteForce(10)

# the time complexity is dependent on the 'height' of the tree, if the fibnoccai sequence you are trying to calculate is the 5th, the height will be roughlt 5 too
# O(2^N)

# we have to always re compute the same sub tree
```

```python
# Memoization/ Top Down Dynamic Programming

# if we are creating the same decision tree, we can store that somewhere
# so that when we are trying to solve the same sub problem, we wont resolve it again

# the dict of the cache contains n as the key and the value of the solution, fn(n)
def memoisation(n, cache):

    if n<=1:
        return n
    
    # checking if it is in the cache
    if n in cache:
        return cache[n]
    
    # back track and save it in the cache
    cache[n] = memoisation(n-1, cache) + memoisation(n-2, cache)

    return cache[n]

memoisation(10, {})

# however space complexity is scaling because of the cache
```

```python
# Bottom Up Dynamic Programming/ True Dynamic Programming Approach

def dp(n):

    if n<1:
        return n
    
    # the sub array to storing the answer

    # index-0, index-1
    dp = [0, 1]

    # now we are trying to solve index-2
    i = 2

    # now we move from 2 to n
    while i<=n:

        # we just need the 2 previous values to calculate

        # dp[1] will be shifted to dp[0]
        # dp[1] will be calculated using the previouse dp[0] and dp[1]
        dp[0], dp[1] = dp[1], dp[0]+dp[1]

        i+=1

    return dp[1]

dp(10)

# time complexity of O(n)
# space complexity of O(2), it is not scaling
```
