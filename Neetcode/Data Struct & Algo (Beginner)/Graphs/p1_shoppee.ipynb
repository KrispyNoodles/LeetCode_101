# Shoppee Xpress Delivery

```python
# shopee xpress delivery

from typing import List
from collections import defaultdict

class Solution:
    
    # have multiple black holes, as long as it is non-zero
    # M is row
    # N is col
    def shopee_xpress(self, M: int, N:int, map: List[List[int]]) -> int:

        # start is always top left
        # end is always bottom right
        
        # total possible path
        answer = []

        visited = set()

        blachole_dict = defaultdict(list)

        # preprocessing all of the blackholes
        for r in range(M):
            for c in range(N):
   
                if map[r][c] != 0:

                    # the key is the value of the blackhole, and it appends the points, of which the blachole appears
                    blachole_dict[map[r][c]].append((r,c))

        def helperFunction(r, c, visited, step_counter):

            # checking for OOB
            if min(r,c)<0 or r==M or c==N or (r,c) in visited:
                return 
            
            # checking if it reached the end, append this answer into the master_set
            if r==M-1 and c==N-1:
                answer.append(step_counter)
                return
            
            # # checking if the step_counter is more than the min(answer), if so can stop early and return
            if len(answer)>=1 and step_counter >= min(answer):
                return
            
            # add the current point
            visited.add((r,c))
            
            # checking if it reaches a portal
            if map[r][c] != 0:
                
                # name of current blachole
                curr_blackhole = map[r][c]

                # finding which new point it goes to
                for new_r, new_c in blachole_dict[curr_blackhole]:

                    # if the points not the same, go and continue moving
                    if (new_r, new_c) != (r,c):
                        helperFunction(new_r, new_c, visited, step_counter)

            # explore different option
            helperFunction(r+1, c, visited, step_counter+1)
            helperFunction(r-1, c, visited, step_counter+1)
            helperFunction(r, c+1, visited, step_counter+1)
            helperFunction(r, c-1, visited, step_counter+1)

            # backtrack
            visited.remove((r,c))

        helperFunction(0, 0, visited, 0)

        if answer == []:
            return -1
        else:
            return min(answer)
```

```python
# shopee xpress delivery
# using BFS to solve now

from typing import List
from collections import defaultdict, deque

class Solution:

    # trying to find all the blacholes
    def blackHole(self, map):

        blackHoleDict = defaultdict(list)

        for r in range(len(map)):
            for c in range(len(map[0])):
                
                # if it is not a 0, it is a BH
                if map[r][c] != 0:

                    # blackhole number
                    bh_number = map[r][c]

                    # adding the point into the blackhole number
                    blackHoleDict[bh_number].append((r,c))

        return blackHoleDict
    
    # have multiple black holes, as long as it is non-zero
    # M is row
    # N is col
    def shopee_xpress(self, M: int, N:int, map: List[List[int]]) -> int:

        blackHoleDict = self.blackHole(map)

        # starting from the top left
        start = (0,0)

        visited = set()
        queue = deque()

        # add start point into the visit and queue
        visited.add(start)
        queue.append(start)

        # initalising steps, steps is counted only when we move to the next "layer"
        steps = 0

        while queue:
            
            for _ in range(len(queue)):

                # pop_ed point
                r, c = queue.popleft()

                # checking if it has reached the destination
                if r==M-1 and c==N-1:
                    print('destination reached')
                    return steps
                
                ## THIS PART DOES NOT WORK BECAUSE IT ADDS A STEP TO THE BH
                # checking if it reaches a BH
                if map[r][c] != 0:

                    # move the new r,c into a new point
                    for new_r, new_c in blackHoleDict[map[r][c]]:
                        
                        # if the new point is not the same as the current point
                        if new_r != r or new_c != c:
                            
                            # checking that it hs not been visited before
                            if (new_r, new_c) not in visited:

                                # append this new teleported into the queue and visisted
                                visited.add((new_r, new_c))
                                queue.append((new_r, new_c))


                # movements
                movements = [[0,1],[1,0],[0,-1],[-1,0]]

                for dr, dc in movements:
                    new_r, new_c = dr+r, dc+c

                    # checking if these new points are OOB or visited before
                    if min(new_r, new_c)<0 or new_r==M or new_c==N or (new_r, new_c) in visited:
                        continue
                        
                    # else add them into the queue and visited
                    visited.add((new_r, new_c))
                    queue.append((new_r, new_c))

            # increment steps for each time it moves to the next "layer"
            steps+=1
        
        # if cant find the end
        return -1
```
