# DFS

Q: Count the unique paths from the top left to the bottom right. A single path may only move along 0's and can’t visit the same cell more than once.

```python
# visit is a hashset containing which points which we had visited before
def dfs(grid, r, c, visit):

    # getting the dimension of the grid
    ROWS, COLS = len(grid), len(grid[0])

    # row and col can get out of bound
    # can use min to check if either is OOB (moving to the negative side)
    # this means that we did not find a path, so we can just return 0
    if (min(r,c))<0:
        return 0

    # if the r,c is moving out of the rows and cols
    if r==ROWS or c==COLS:
        return 0

    # or if we reach to a blocked position
    if grid[r][c]==1:
        return 0

    # lastly, if we reach a position that we have reached before
    if (r,c) in visit:
        return 0


    # if we reach the bottom corner, we have reached the end
    if r == ROWS-1 and c == COLS-1:
        return 1

    # now we can select the option using visit
    visit.add((r,c))

    # initalise counter:
    count = 0

    # explore other option
    count += dfs(grid, r+1, c, visit)
    count += dfs(grid, r-1, c, visit)
    count += dfs(grid, r, c+1, visit)
    count += dfs(grid, r, c-1, visit)

    # backtrack
    visit.remove((r,c))

    return count

grid =[[]]

answer = dfs(grid,0,0,set())

```
