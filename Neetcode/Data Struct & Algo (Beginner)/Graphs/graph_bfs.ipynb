# BFS

Q: Find the length of the shortest path from the top left of the grid to the bottom right

O(n*m)

```python
from collections import deque

# Shortest path from top left to bottom right
def bfs(grid):
    ROWS, COLS = len(grid), len(grid[0])

    visited = set()
    queue = deque()

    # add in the first top left
    visited.add((0,0))
    queue.append((0,0))

    length = 0

    while queue:

        for _ in range(len(queue)):
            
            # pop left, to show that it has been used from the queue
            r,c = queue.popleft()

            # checking if you are already at the destination
            if r==ROWS-1 and c==COLS-1:
                return length
        
            neighbours = [[1,0], [-1,0], [0,1], [0,-1]]

            # difference in rows and colums
            for dr, dc in neighbours:
                
                # these are choosing, when it should skip adding the points into the queue and visited
                # checking if they are OOB
                if min(r+dr,r+dc)<0 or (r+dr)==ROWS or (r+dc)==COLS:
                    continue

                # checking if they have been visited before or they cannot be entered into because they are 1's
                if (r+dr, c+dc) in visited or grid[r+dr][r+dc] == 1: 
                    continue

                # add in the point of the new direction
                queue.append((r+dr, c+dc))
                visited.add((r+dr, c+dc))

        # each time we repeate the queue, we are moving to the next layer
        # from the perspective of the next layer on the map, it is referring to after taking a step later, where can we go next and add it to our queue
        length+=1

# time complexity of O(n*m) where n is the row and columns, meaning we will only visit each cell once
```
