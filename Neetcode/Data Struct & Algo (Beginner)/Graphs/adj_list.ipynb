```python
class GraphNode:
    def __init__(self, val):
        self.val = val
        self.neighbours = []

# or use a HshaMap
adjList = {"A": [], "B": []}

# given the directed edges, build an adjacent list
# [A,B] means A points to B
edges = [["A", "B"], ["B", "C"], ["B", "E"], ["C", "E"], ["E", "D"]]

from collections import defaultdict

# declare empty
adjList = defaultdict(list)

for src, dst in edges:
    # if source not in dict
    if src not in adjList:
        adjList[src]=[]

    # if destination not in dict
    if dst not in adjList:
        adjList[dst]=[]

    adjList[src].append(dst)

adjList
```

```python
# not as much edge cases as matrix
# count paths (backtracking)
def dfs(node, target, adjList, visit):

    # if the node has been visited before
    if node in visit:
        return 0

    # if the node is the target
    if node==target:
        return 1

    # addint the current node into the visit
    count=0
    visit.add(node)

    # the adjList contains the neighbours of the node
    for neighbour in adjList[node]:
        count += dfs(neighbour, target, adjList, visit)
    
    # backtrack
    visit.remove(node)

    return count

# the nubmer of paths from A to E
print(dfs("A","E",adjList, set()))

# time complexity of O(V^2) where N==V and N is the number of edges
```

```python
from collections import deque

def bfs(node, target, adjList):

    length = 0
    visited = set()
    queue = deque()

    # add the tarting node to the queue and the visit set
    queue.append(node)
    visited.add(node)

    while queue:

        for _ in range(len(queue)):

            curr = queue.popleft()

            # checking if it is the target
            if curr == target:
                return length

            # similar to matrix
            for neighbour in adjList[curr]:

                # checking if the neighbours had been visited before
                if neighbour not in visited:

                    # add it to the queue and visited
                    queue.append(neighbour)
                    visited.add(neighbour)

        # else add length+1, for each time the queue is ran
        length+=1
    
    return length

print(bfs("A","E",adjList))

# time complexit is O(V+E) where V is the number of Nodes and E is the number of Edges
```
