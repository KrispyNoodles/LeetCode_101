# Heap Properties

A heap is a specialized, tree-based data strcture. It uses a concept called the Priority Queue, but sometimes these two terms 'Heap' and 'Priority Queue' are used interchangeably. Before this chapter, we had learned that queues opereate with a firs-in-first (FIFO) out basis but for Priority Queues, each element has a different priority and teh element with the highest or the lowest priority can be removed first, regardless of the order in which it was added.


A complete tree: meaning every layer has to be filled except the leaf nodes
Duplicated values allowed

For Min Heap:
For every node's, its decendants have to be greater or equal to the root node (meaning the smallest is above)

For Max Heap:
For every node's, its decendants have to be smaller or equal to the root node (meaning the largest is above)

Structure in the array:
root->left->right

leftChild = 2*i
rightChild = 2*i+1
parent = i/2

**where i is the index of the root node

## Pushing into a min heap

1. check if current location is bigger than parent?
2a. if it is not then swap
4. repeat till it reaches step 2b

2b. if it is then it is at the correct place


```python
class Heap:
    def __init__(self):

        # initialising the heap
        self.heap = [0]
    

    def push(self, val):
        
        self.heap.append(val)
        i = len(self.heap)-1 ## -1 because of the 1 index

        # Percolate up
        # as long as the value that is currently added, is smaller than the parent
        # we execute this
        while self.heap[i] < self.heap[i//2]:

            # doing the swap
            self.heap[i//2], self.heap[i] = self.heap[i], self.heap[i//2]

            # updating the current index of the value that was just added
            i = i//2

    # we are replacing the last element with the element that is being poped, at the root (to ensure structure integrity)
    # then we are checking if it fulfils the order property and if it doesnt, we Percolate it down, similar to how a value is added

    # leftChild = 2*i
    # rightChild = 2*i+1
    # parent = i/2

    # element that is being pop, cant be decided and it is immediately selected to be the root that is to be pop-ed
    def pop(self):

        # there is no element within
        if len(self.heap)==1:
            return None
        
        # there is only 1 element
        if len(self.heap)==2:
            return self.heap.pop()
        
        # element to be removed
        res = self.heap[1]

        # moving the last value to the root
        self.heap[1] = self.heap.pop()
        
        # pointer at the root node, then percolate down
        i = 1

        # as long as the child is within the length of the heap, we can run the loop
        while 2*i < len(self.heap):

            # it could have both children, no children or only left (no such thing as no right, because it populates left then right first)
            
            # we are replacing the root node with the right child
            # do we have a right child AND right child smaller than left child AND the current node we are at is greater than the right child
            if (2*i+1 < len(self.heap)) and (self.heap[2*i]>self.heap[2*i+1]) and (self.heap[i]>self.heap[2*i+1]):

                self.heap[i],self.heap[2*i+1] = self.heap[2*i+1],self.heap[i]

                # update i
                i = 2*i+1

            # we try swapping the root with the left child instead
            elif (self.heap[i]>self.heap[2*i]):

                self.heap[i],self.heap[2*i] = self.heap[2*i],self.heap[i]

                # update i
                i = 2*i

            else:
                # it has finally found the correct place, no swap needed
                break

        return res

# time complexity of pop and push is log(n) because of the height of a balanced tree       
```

```python
class Heap:

    # creating heapify
    def heapift(self, arr):

        # 0th position is move to the end
        arr.append(arr[0])

        self.heap = arr

        curr = (len(self.heap)-1)//2

        while curr>0:

            # perculate down
            i = curr
            
            # as long as the child is within the length of the heap, we can run the loop
            while 2*i < len(self.heap):

                # it could have both children, no children or only left (no such thing as no right, because it populates left then right first)
                
                # we are replacing the root node with the right child
                # do we have a right child AND right child smaller than left child AND the current node we are at is greater than the right child
                if (2*i+1 < len(self.heap)) and (self.heap[2*i]>self.heap[2*i+1]) and (self.heap[i]>self.heap[2*i+1]):

                    self.heap[i],self.heap[2*i+1] = self.heap[2*i+1],self.heap[i]

                    # update i
                    i = 2*i+1

                # we try swapping the root with the left child instead
                elif (self.heap[i]>self.heap[2*i]):

                    self.heap[i],self.heap[2*i] = self.heap[2*i],self.heap[i]

                    # update i
                    i = 2*i

                else:
                    # it has finally found the correct place, no swap needed
                    break
```
