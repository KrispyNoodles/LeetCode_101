# 215. Kth Largest Element in an Array

```python
# solving using heaps
import heapq
from typing import List

class Solution:
    def findKthLargest(self, nums: List[int], k: int) -> int:

        new_list = []

        # convert to negative
        for i in nums:
            new_list.append(-i)
        
        # creating a heap
        heapq.heapify(new_list)

        # poping till there are k elements inside, then return the "smallest"
        for i in range(k):
            answer = heapq.heappop(new_list)

        # putting negative back
        return -answer

# time complexity of O(n+klog(n)) because of creating a heap

# O(n) from covnerting to negatives, each pop uses O(log(n)) and it runs k times 

# n is the number of elements in the input list nums
# k is the index of the largest element to be found
```

```python
import heapq

class Solution:
    def findKthLargest(self, nums, k):
        
        # this creats a heap internally and returns a list of the k largest elements, sorted in descending order
        print(heapq.nlargest(k,nums))

        return heapq.nlargest(k, nums)[-1]

# time complexity of O(nlog(n)) because of creating a heap
```

```python
# Input: 
nums = [3,2,3,1,2,4,5,5,6]
k = 4

# Output: 4

sol = Solution()
sol.findKthLargest(nums, k)
```
