# 1046. Last Stone Weight

```python
import heapq
from typing import List

class Solution:
    def lastStoneWeight(self, stones: List[int]) -> int:

        # because the heapq is a min_heap, need to convert the stones into negative
        new_stones = []

        for stone in stones:
            new_stones.append(-stone)

        # creating a heap from the stones
        heapq.heapify(new_stones)

        while len(new_stones)>1:

            # pop the biggest 2 stones
            stone_1 = heapq.heappop(new_stones)
            stone_2 =heapq.heappop(new_stones)

            # minus the big from the small stone
            if stone_1==stone_2:
                print('same stone')
            else:
                new_stone = abs(stone_1-stone_2)

                heapq.heappush(new_stones, -new_stone)
        
        if new_stones == []:
            return 0

        else:
            return abs(new_stones[0])
```
