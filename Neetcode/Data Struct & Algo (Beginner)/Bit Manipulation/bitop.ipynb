# BIT MANIPULATION

```python
# AND Operation
n = 1 & 1

# OR
n = 1 | 1

# XOR (only one has to be 1)
n = 0 ^ 1

# NOT (negation)
n = ~n

# Bit shifting
n = 1

# shift left
n = n<<1
# shift right
n = n>>1
```

```python
class Solution:
    def getSum(self, a: int, b: int) -> int:
        
        carry = 0
        result = 0

        for i in range(32):
            
            # checking the value at each position
            # from right most to left
            a_bit = (a>>i)&1
            b_bit = (b>>i)&1

            if carry == 0:
                # checking for carry
                if a_bit==1 and b_bit==1:
                    carry = 1

                # either is 1
                elif a_bit==1 or b_bit==1:

                    # adding the bit to the correct position
                    # shifting the 1 to the left by i position
                    result = result | 1<<i

                else:
                    # let the for loop increment
                    print('just shift')
                    
            else:
                # checking for carry
                if a_bit==1 and b_bit==1:
                    carry = 1

                    result = result | 1<<i

                # either is 1
                elif a_bit==1 or b_bit==1:
                    carry = 1
                
                # both is 0
                else:
                    result = result | 1<<i

                    #reset carry 
                    carry=0
        
        # i have zero idea LOL
        if result > 0x7FFFFFFF:
            result = ~(result^0xFFFFFFFF)

        return result

# time complexity O(1)
# space complexity O(1)
```

```python
# counting Bits
# int to bits

# counting the number of 1 bits
def countBits(n):

    count = 0

    while n>0:

        # as lonng as the smallest bit is a 1, the count will be increment
        if n&1==1:
            count+=1

        # shift to the right by 1, after counting
        n = n >>1
    return count

# keep shifting to the right by 1
# 23 = 10111
# 11 = 1011
# 5 = 101
# 2 = 10
# 1 = 1
countBits(23)
```
