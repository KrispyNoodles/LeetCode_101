# 283. Move Zeroes

```python
from typing import List

class Solution:
    def moveZeroes(self, nums: List[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        
        # shifting the zero to the end
        left = 0

        for i in range(len(nums)):

            if nums[i] != 0:
                nums[left] = nums[i]
                left+=1
        
        # where left ends, we add the 0 now
        while len(nums)>left:
            nums[left]=0
            left+=1
        
        print(nums)
```
