# Two Pointers

sliding window is a subset of a two pointer

sliding windows cares about the values within the two poitns (the range) where as the two pointer we only care about the two individual points

```python
# compare the pointer at the left and right, 
# if they are equal 
# then increase the left and decrease the right

# if they are not equal, it stops and it is not a palindrone

# if the right is lesser than the left, it means that they have crossed path and we have compared all values

# Q: valid palidrone?
def isPalidrone(word):

    left, right = 0, len(word)-1

    # this ensures that it stops when the points meet or when they cross
    while right>left:

        # do the check
        if word[left] != word[right]:
            return False

        left+=1
        right-=1
    
    # after everything return True
    return True

isPalidrone('catac')
```

```python
# Q: Given a sorted input array, return the two indices of two elements which sum up to the target value. Assume there's exactly one solution.

# brute force method O(n^2) because we find one point, move the other pointer to check we can find the value
# then move the first point to the end



# the left pointer is the smallest value and right pointer is the largest value

# when we sum both pointers and it is still bigger than the target
# we can safely assume that the right pointer is useless as we will never be able to use it in our answer to find the target
# then we shift the right pointer to the left

# on the other hand, when the sum is smaller than the target, 
# we can see it again that there is no possible value that we can add to the left pointer that can give us our answer
# we are not sure if the right pointer is correct, so we shift the left pointer to the right

# if both pointer adds to the sum, we have found the target

def targetSum(nums, target):

    left, right = 0, len(nums)-1
    
    # we know we gonna find a solution, so we dont needa create a return out of the while loop
    while right>left:

        # checking if they meet the target
        sum = nums[left] + nums[right]

        if sum == target:
            return [left, right]
        
        # else check if sum is bigger or smaller than target to know which pointer to move
        elif sum>target:
            right-=1

        else:
            left+=1

targetSum([-1,2,3,4,8,9], 7)
```
