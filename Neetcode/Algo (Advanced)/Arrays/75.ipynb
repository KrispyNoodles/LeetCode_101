# 75. Sort Colors

```python
from typing import List

class Solution:
    def sortColors(self, nums: List[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        
        # left writer
        left = 0

        # bring all 0 in front
        for i in range(len(nums)):
            
            # for 0 values
            if nums[i] == 0:
                # swaperoo
                nums[left], nums[i] = nums[i], nums[left]
                left+=1

        # find where one left off
        one_writer = left
        for i in range(one_writer, len(nums)):
            if nums[i] == 1:
                # swaperoo
                nums[one_writer], nums[i] = nums[i], nums[one_writer]
                one_writer+=1

        # find where two left off
        two_writer = one_writer
        for i in range(two_writer, len(nums)):
            if nums[i] == 2:
                # swaperoo
                nums[two_writer], nums[i] = nums[i], nums[two_writer]
                two_writer+=1

# time complexity of O(n)
# space of O(1)
```

```python
class Solution:
    def sortColors(self, nums: List[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        
        # low points to the next spot for 0
        low = 0

        curr = 0

        # high points to the next spot for 2
        high = len(nums)-1

        # curr must be below high, because high is where all the 2's value starts
        while curr<=high:
            
            # for 0 values
            if nums[curr] == 0:
                # swaperoo
                nums[low], nums[curr] = nums[curr], nums[low]

                # move to the next curr
                low+=1
                curr+=1

            # for 2 values
            elif nums[curr] == 2:
                # swaperoo
                # dont have to move the curr as still need to
                # check what to do with the curr elemnt
                nums[high], nums[curr] = nums[curr], nums[high]
                high-=1

            else:
                curr+=1

# time complexity of O(n)
# space of O(1)
```
