```python

````python
# We use prefix to prevent repeated work
# A classic quesiton is "Given an array of values, design a data structure that can query the sum of subarray of the values

array_1 = [2,-1,3,-3,4]

# creating preifx

def querySum(l,r):

    # creating prefix sum array with the first element in it
    prefix_arr = [array_1[0]]
    
    for i in range(1, len(array_1)):

        # sum and then add into the prefix_arr
        prefix_arr.append(prefix_arr[-1]+array_1[i])

    print(prefix_arr)
    print(array_1)

    # because we want the sum of values range from the right to the left (including the left)
    sum_val = prefix_arr[r]-prefix_arr[l-1]

    return sum_val

querySum(1,3)

## 6min 53sec
````

````python
# Creating PrefixSum:

class PrefixSum:
    
    def __init__(self, nums):
        self.prefix = []
        total = 0

        for n in nums:
            total+=n
            self.prefix.append(total)
    
    def rangeSum(self, left, right):

        preRight = self.prefix[right]
        preLeft = self.prefix[left-1] if left > 0 else 0

        return (preRight-preLeft)


array_1 = [2,-1,3,-3,4]

object_1 = PrefixSum(array_1)
object_1.rangeSum(1,3)
````
