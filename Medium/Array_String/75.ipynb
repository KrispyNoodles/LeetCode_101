# 75. Sort Colors

```python
from typing import List

# merge sort
class Solution:
    def sortColors(self, nums: List[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        # do merge sort
        self.merge_sort(nums, 0, len(nums)-1)
    
    def merge_sort(self, nums, left, right):

        # if right == left
        # single element
        if right == left:
            return nums
        
        # mid calculation
        mid = (left + right)//2

        # solve for left and right
        self.merge_sort(nums, left, mid)
        self.merge_sort(nums, mid+1, right)

        self.merge(nums, left, mid, right)

        return nums
    
    def merge(self, nums, left, mid, right):

        # left array and right array
        left_array = nums[left:mid+1]
        right_array = nums[mid+1:right+1]

        # index to compare
        # i and j for left and right
        i,j = 0,0

        # k for the index that it is populating in
        k = left

        while i < len(left_array) and j < len(right_array):

            # add in j
            if left_array[i] >= right_array[j]:
                nums[k] = right_array[j]
                # shift j
                j += 1
            else:
                nums[k] = left_array[i]
                i += 1

            # increment k
            k+=1
        
        # add in the ends
        while i < len(left_array):
            nums[k] = left_array[i]
            i += 1
            k += 1

        while j < len(right_array):
            nums[k] = right_array[j]
            j += 1
            k += 1

# time complexity of O(nlog(n))
```

```python
from typing import List

# insertion sort
class Solution:

    # trying to do increasing
    def sortColors(self, nums: List[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """

        for i in range(1, len(nums)):

            j = i-1

            # put the smallest on the left
            while j>=0 and nums[j] >= nums[j+1]:

                # do the swap
                nums[j], nums[j+1] = nums[j+1], nums[j] 

                # decrement j
                j-=1

# time complexity of O(n^2)
```

```python
from typing import List
from collections import defaultdict

# create dictionary
class Solution:

    # trying to do increasing
    def sortColors(self, nums: List[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """

        count_dict = defaultdict(int)

        for value in nums:
            count_dict[value]+=1

        # sort based of key
        count_dict = sorted(count_dict.items(), key=lambda x: x[0])

        index_counter = 0 

        for key, value in count_dict:            
            for _ in range(value):

                nums[index_counter] = key
                index_counter += 1

# time complecity of O(n + klog(k))
# where n is the number of elements from creating the dict and k the number of unique keys
```

```python
from typing import List

# bucket sort
class Solution:
    def sortColors(self, nums: List[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        
        # because it is either 0,1 or 2
        bucket_arr = [0, 0, 0]

        for value in nums:
            bucket_arr[value] += 1

        index = 0
        
        for values in range(len(bucket_arr)):
            for _ in range(bucket_arr[values]):

                nums[index] = values
                index += 1
    
        return nums
```

```python
# Input: 
nums = [2,0,2,1,1,0]

sol = Solution()
sol.sortColors(nums)

# Output: [0,0,1,1,2,2]
```
