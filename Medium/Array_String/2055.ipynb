```python
from typing import List

# TLE Time Limit Excceeded
class Solution:
    def platesBetweenCandles(self, s: str, queries: List[List[int]]) -> List[int]:
        
        # answer
        answer = []

        for query in queries:
            conter = 0    
            new_problem = s[query[0]:query[1]+1]

            if '|' not in new_problem:
                answer.append(0)
                continue

            # finding left boundary and right boundary
            index_left = new_problem.index('|')

            reverse_problem = new_problem[::-1]
            index_right = reverse_problem.index('|')
            index_right = len(new_problem)-1-index_right

            for character in new_problem[index_left:index_right+1]:
                if character == '*':
                    conter+=1
            
            answer.append(conter)

        return answer
```
