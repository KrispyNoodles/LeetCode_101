```python
# using Binary search
from typing import List

# cheating using rindex and index
class Solution:
    def searchRange(self, nums: List[int], target: int) -> List[int]:

        if target not in nums:
            return [-1, -1]
            
        else:
            # do bin search on current array
            left = self.left_binarySearch(nums, target)

            # and do bin search on flip array
            right = self.right_binarySearch(nums, target)
            
            return [left, right]
    
    def left_binarySearch(self, nums, target):

        left = 0
        right = len(nums)-1

        while right>=left:

            middle = (right+left)//2

            # this means that the target is on the left side
            if nums[middle]>=target:
                right = middle-1

            else:
                left = middle+1

        return left

    def right_binarySearch(self, nums, target):

        left = 0
        right = len(nums)-1

        while right>=left:

            middle = (right+left)//2

            # this means that the target is on the left side
            if nums[middle]<=target:
                left = middle+1

            else:
                right = middle-1

        return right
```

```python
# Input: 
nums = [5,7,7,8,8,10]
target = 8

sol = Solution()
sol.right_binarySearch(nums, target)

# Output: 3
```
