# 215. Kth Largest Element in an Array

```python
from typing import List

class Solution:
    def findKthLargest(self, nums: List[int], k: int) -> int:
        
        nums.sort()

        return nums[-k]
```

```python
# without using sort library
# TLE exceeded because I am sorting the whole array which isnt neccesary
from typing import List

class Solution:
    def findKthLargest(self, nums: List[int], k: int) -> int:
        
        # implement quick sort
        s = 0
        e = len(nums)-1

        nums = self.quickSort(nums,s,e)

        return nums[-k]
    
    def quickSort(self, nums, s, e):
 
        # for cases where s>e
        if s >= e:
            return nums
        
        pivot = nums[e]
        left = s

        # compare all the values from the first index tot the second last, beside the pivot
        for i in range(s,e):

            if pivot > nums[i]:

                nums[i], nums[left] = nums[left], nums[i]
                
                # increment left
                left += 1

        # after everthing move the pivot to the correct spot
        nums[e], nums[left] = nums[left], nums[e]

        # since the pivot value is placed in the left so...
        # solve for the left and right of the pivot
        self.quickSort(nums, s, left-1)
        self.quickSort(nums, left+1, e)

        return nums
```

```python
# without using sort library
# TLE exceeded because I am sorting the whole array which isnt neccesary

# still TLE after optimisation.... LOLOLS
from typing import List

class Solution:
    def findKthLargest(self, nums: List[int], k: int) -> int:
        
        # implement quick sort
        s = 0
        e = len(nums)-1

        # finding the target index
        target = len(nums) - k

        largest_value = self.quickSort(nums,s,e,target)

        return largest_value
    
    def quickSort(self, nums, s, e, target):
 
        # for cases where s>e
        if s >= e:
            return nums
        
        pivot = nums[e]
        left = s

        # compare all the values from the first index tot the second last, beside the pivot
        for i in range(s,e):

            if pivot > nums[i]:

                nums[i], nums[left] = nums[left], nums[i]
                
                # increment left
                left += 1

        # after everthing move the pivot to the correct spot
        nums[e], nums[left] = nums[left], nums[e]

        # if the target is found
        if left == target:
            return nums[left]
        
        # it means that it is on the right side
        elif left < target:
            return self.quickSort(nums, left+1, e, target)
        
        # else just solve the left side
        else:
            # since the pivot value is placed in the left so...
            # solve for the left and right of the pivot
            return self.quickSort(nums, s, left-1, target)
```

```python
# try quick sort instead
# still TLE, LOL
from typing import List

class Solution:
    def findKthLargest(self, nums: List[int], k: int) -> int:

        # quick sort the nums array
        for i in range(1, len(nums)):
            j = i-1

            # put the biggest on the left
            while j >= 0 and nums[j] <= nums[j+1]:
                nums[j], nums[j+1] = nums[j+1], nums[j]

                # decrement j
                j -= 1
        
        # because it is zero index so...
        return nums[k-1]
```

```python
# using dict instead
from typing import List
from collections import defaultdict

class Solution:
    def findKthLargest(self, nums: List[int], k: int) -> int:

        # creating a dict
        num_dict = defaultdict(int)

        for value in nums:
            num_dict[value] += 1

        # sort it from largest to smallest
        num_dict = sorted(num_dict.items(), key=lambda x: x[0], reverse=True)

        index_coutner = 0

        for value, count in num_dict:
            
            index_coutner += count

            if index_coutner >= k:
                return value
            
            
```

```python
# Input: 
nums = [3,2,3,1,2,4,5,5,6]
k = 4

# Output: 4

sol = Solution()
sol.findKthLargest(nums, k)
```
