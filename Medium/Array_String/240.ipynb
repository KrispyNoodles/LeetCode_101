```python
from typing import List

class Solution:
    def searchMatrix(self, matrix: List[List[int]], target: int) -> bool:

        for arrays in matrix:
            if target in arrays:
                return True

        return False

# time complexity of O(n*m) because it is brute
```

```python
# bin search
from typing import List

class Solution:
    def searchMatrix(self, matrix: List[List[int]], target: int) -> bool:

        for arrays in matrix:
            answer = self.bin_search(arrays, target)

            if answer:
                return True
            
        return False
    
    def bin_search(self, array_1, target):

        left = 0
        right = len(array_1)-1

        while right>=left:

            middle = (right+left)//2

            if array_1[middle] > target:
                # value is on the left
                right = middle-1

            elif array_1[middle] < target:
                # value is on the left
                left = middle+1

            else:
                return True
        
        return False
    
# time complexity of O(n*log(m)) because it is bin search in each row
```

```python
# moving based on how the pattern of the matrix looks like

from typing import List

class Solution:
    def searchMatrix(self, matrix: List[List[int]], target: int) -> bool:

        matrix_rows = len(matrix)
        matrix_cols = len(matrix[0])

        # initialisation by starting on the bottom left
        # it moves up if the target is smaller and move right if it is bigger
        
        # pointer for rows
        i = matrix_rows-1

        # pointer for cols
        j = 0

        while i>=0 and j<matrix_cols:

            if matrix[i][j] == target:
                return True
            
            # if the value is smaller, can go up by decreasing i
            elif matrix[i][j] > target:
                i -= 1
            
            # if the value is larger, can increase the j
            else:
                j += 1
        
        return False

```

```python
# Input: 
matrix = [[1,4,7,11,15],[2,5,8,12,19],[3,6,9,16,22],[10,13,14,17,24],[18,21,23,26,30]]
target = 5
# Output: true

sol = Solution()
sol.searchMatrix(matrix, target)
```
