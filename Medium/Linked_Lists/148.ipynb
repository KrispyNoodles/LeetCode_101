```python
from typing import Optional

# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def sortList(self, head: Optional[ListNode]) -> Optional[ListNode]:

        temp_array = []

        # cheese strategy, retrieve the elemtns into array sort then put back into list node
        while head != None:
            temp_array.append(head.val)
            head = head.next

        # sort it
        temp_array.sort()

        # put back into list now
        dummy_node = ListNode(-1)
        tail = dummy_node

        for value in temp_array:
            # create node
            tail.next = ListNode(value)

            # reassign tail
            tail = tail.next

        return dummy_node.next


    def view_link_list(self, head: Optional[ListNode]) -> Optional[ListNode]:

        print(head.val)

        curr = head

        while curr.next != None:

            curr = curr.next
            print(curr.val)
```

```python
# doing with proper algo of merge sort

from typing import Optional

# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
    
class Solution:
    def sortList(self, head: Optional[ListNode]) -> Optional[ListNode]:

        # for cases with no head
        # and if it is only one element
        if not head or not head.next:
            return head
        
        left = head
        mid = self.getmid(head)
        right = mid.next

        # remove the mid.next to create two seperate lists
        mid.next = None

        # solve for left
        left = self.sortList(left)

        # solve for right
        right = self.sortList(right)

        # merge both ascending
        return self.mergeTwoLists(left, right)


    def getmid(self, head):
        # the idea is that we have two pointers
        # and for each time the slow moves one the fast moves twice, 
        # therefore by the time the fast reaches the end, the slow would reach the centre

        slow = head
        fast = head.next

        while fast and fast.next != None:

            # move once
            slow = slow.next
            
            # move two times
            fast = fast.next.next

        return slow
    
    def mergeTwoLists(self, list1: Optional[ListNode], list2: Optional[ListNode]) -> Optional[ListNode]:

        dummy_node = ListNode(-1)
        tail = dummy_node

        # as long as list1 and list2 are not empty
        while list1 and list2:  

            # checking if which value is bigger
            if list1.val >= list2.val:

                # list2 becomes the new tail
                tail.next = list2

                # shift list2
                list2 = list2.next
            
            else:
                tail.next = list1
                list1 = list1.next

            # change the new tail
            tail = tail.next
    
        # checking if either is left
        if list1:
            tail.next = list1
        if list2:
            tail.next = list2

        return dummy_node.next
    
    def view_link_list(self, head: Optional[ListNode]) -> Optional[ListNode]:

        print(head.val)

        curr = head

        while curr.next != None:

            curr = curr.next
            print(curr.val)
```
