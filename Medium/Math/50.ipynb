# 50. Pow(x, n)

```python
# TLE tragic....
class Solution:
    def myPow(self, x: float, n: int) -> float:

        # power of 1 is the value
        if n == 1:
            return x
        
        # power of 0 is 1
        elif n == 0:
            return 1
        
        # if it is a negative pow then call the same function but with the value as 1/x and switch the sign of n
        elif n < 0:
            return self.myPow(1/x, -n)
        
        # else it is times x
        else:
            return x * self.myPow(x, n-1)
```

```python
# neetcode soln
class Solution:
    def myPow(self, x: float, n: int) -> float:

        # creating a power function to solve only positive
        def helper(x, n):

            # create the base cases
            if n == 1:
                return x
            
            elif n == 0:
                return 1
            
            # you just do the calulcation for odd or even fist
            res = helper(x, n//2)

            # multiply by itseld
            res = res * res

            # the odd can be adjusted in here
            if n%2==0:
                return res
            else:
                # adding one more x here
                return res*x
            
        result = helper(x, abs(n))

        if n >0:
            return result
        
        else:
            return 1/result
```

```python
# Input: 
x = 2.00000
n = -200000000

sol = Solution()
sol.myPow(x, n)
```

```python
# Input: 
x = 2.00000
n = -2

sol = Solution()
sol.myPow(x, n)

# Output: 0.25000
# Explanation: 2^-2 = 1/(2^2) = 1/4 = 0.25
```
