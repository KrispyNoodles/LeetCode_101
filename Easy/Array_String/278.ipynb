# 278. First Bad Version

```python
# The isBadVersion API is already defined for you.
# def isBadVersion(version: int) -> bool:

# redo
class Solution:
    def firstBadVersion(self, n: int) -> int:

        def helper_function(low, high):

            # answer has been found
            if low>high:
                return low
            
            # check the version
            middle = (high+low)//2

            # check if middle is the correct version
            if self.isBadVersion(middle):
                

                # can afford to move right by 1
                return helper_function(low, middle-1)

            else:
                return helper_function(middle+1, high)

        return helper_function(0,n)

    def isBadVersion(self, num: int):
    
        secret_num = 4

        if num == secret_num:
            return True

        else:
            return False
```
