# 3379. Transformed Array

```python
from typing import List

class Solution:
    def constructTransformedArray(self, nums: List[int]) -> List[int]:
        
        # if i is more than 0 than move right

        # if i less than 0 then move left

        # if i==0

        len_nums = len(nums)

        new_array = []

        for index in range(len_nums):

            value = nums[index]
            val_move = (index+value)%len_nums

            # move right
            if value>0:
                                
                new_array.append(nums[val_move])

            # move left
            # since index+value == index-(-value)
            elif value<0:

                new_array.append(nums[val_move])

            else:
                new_array.append(value)
        
        return new_array[:len_nums]
```

```python
# Input: 
nums = [-1,4,-1]

sol = Solution()
sol.constructTransformedArray(nums)

# Output: [-1,-1,4]
```

```python
# Input: 
nums = [-10]

sol = Solution()
sol.constructTransformedArray(nums)

# Output: [-10]
```

```python
# Input: 
nums = [3,-2,1,1]

sol = Solution()
sol.constructTransformedArray(nums)

# Output: 
# [1,1,1,3]
```
