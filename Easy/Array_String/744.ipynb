# 744. Find Smallest Letter Greater Than Target

```python
from typing import List

# the list is already sorted
class Solution:
    def nextGreatestLetter(self, letters: List[str], target: str) -> str:

        # ord function converts alphabet into number
        target_num = ord(target)

        new_letters = []

        # convert into numbers
        for alphabets in letters:
            new_letters.append(ord(alphabets))

        # then do bin_search
        index = self.bin_Search(new_letters, target_num)

        # checking for index if over the len
        if index > len(new_letters)-1:
            index = 0

        return letters[index]

    def bin_Search(self, array, target):

        left = 0
        right = len(array)-1

        while right >= left:

            middle = (right+left)//2

            # if the character is found move the left to the right of the middle by 1
            if array[middle] == target:
                left = middle + 1

            # this means that the value is on the right side
            elif array[middle] > target:
                right = middle - 1
            
            else:
                left = middle + 1
        
        return left
```

```python
from typing import List

# not converting to numbers, because alphabets can be compared
# calling the binsearch directly

# the list is already sorted
class Solution:
    def nextGreatestLetter(self, letters: List[str], target: str) -> str:
        
        left = 0
        right = len(letters)-1

        while right >= left:

            middle = (right+left)//2

            # if the character is found move the left to the right of the middle by 1
            if letters[middle] == target:
                left = middle + 1

            # this means that the value is on the right side
            elif letters[middle] > target:
                right = middle - 1
            
            else:
                left = middle + 1

        # checking for index if over the len
        if left > len(letters)-1:
            left = 0

        return letters[left]
```

```python
letters = ["e","e","g","g"]
target = "g"

sol = Solution()
sol.nextGreatestLetter(letters, target)

# Explanation: "e"
```

```python
letters = ["c","f","j"]
target = "c"

sol = Solution()
sol.nextGreatestLetter(letters, target)

# Explanation: The smallest character that is lexicographically greater than 'c' in letters is 'f'.
```
