# 3010. Divide an Array Into Subarrays With Minimum Cost I

```python
from typing import List

class Solution:
    def minimumCost(self, nums: List[int]) -> int:
        
        # add in the first value
        value_1 = nums[0]

        # sort the next sectin
        nums = nums[1::]
        nums.sort()

        # find the next smallest elemetn
        value_2 = min(nums)

        # remove from the array and find again
        # remove removes by value rather by index
        nums.remove(value_2)
        value_3 = min(nums)

        return value_1 + value_2 + value_3
```

```python
nums = [1,2,3,12]

sol = Solution()
sol.minimumCost(nums)
```
