# 3637. Trionic Array I

```python
from typing import List

class Solution:
    def isTrionic(self, nums: List[int]) -> bool:
    
        n = len(nums)
        i = 1

        # must check is the length more than 4 to be working
        if n<4:
            return False

        # declaraiton
        final_check = 0

        # need to check if strictly increasing
        # nums[i-1] < nums[i], i at 1 this is better as it does not move out of the index

        # nums[i] < nums[i+1], i at 0
        while i < n and nums[i-1] < nums[i]:
            i+=1

        # declaration of p
        p = i-1
        
        # checking for second strictly decrease
        while i < n and nums[i-1] > nums[i]:
            i+=1

        q = i-1

        # last check
        # then check for strictly increasing again
        while i < n and nums[i-1] < nums[i]:
            i+=1

            # keep updating the n
            final_check = i-1

        # if p>0, 
        if p>0 and p<q and final_check==n-1:
            return True
        
        else:
            return False
```
