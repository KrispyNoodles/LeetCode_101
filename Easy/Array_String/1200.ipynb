# 1200. Minimum Absolute Difference

```python
from typing import List

class Solution:
    def minimumAbsDifference(self, arr: List[int]) -> List[List[int]]:

        index = 0
        difference_val = None

        # sort the array from small to big
        arr.sort()
        
        # finding the smallest difference first
        while index <= len(arr)-2:

            temp_arr = arr[index:index+2]

            current_difference_val = temp_arr[1]-temp_arr[0]

            # getting the difference value
            if difference_val is None:
                difference_val = current_difference_val
            else:
                difference_val = min(difference_val, current_difference_val)

            index+=1

        answer = []
        index = 0

        # then getting the answer now
        while index <= len(arr)-2:

            temp_arr = arr[index:index+2]

            current_difference_val = temp_arr[1]-temp_arr[0]

            if current_difference_val == difference_val:
                answer.append(temp_arr)
            index+=1

        return answer

# time complexit of O(nlog(n)) from sorting
# space complexity of O(n)
```

```python
# Input: 
arr = [3,8,-10,23,19,-4,-14,27]

sol = Solution()
sol.minimumAbsDifference(arr)

# Output: [[-14,-10],[19,23],[23,27]]
```
