# 1984. Minimum Difference Between Highest and Lowest of K Scores

```python
from typing import List

class Solution:
    def minimumDifference(self, nums: List[int], k: int) -> int:

        i = 0
        current_minimum = None

        # sort first
        nums.sort()
        
        # shift the window
        while i <= len(nums)-k:
            
            # new_arr is the window
            new_arr = nums[i:i+k]

            # find the min and the max difference of each new_arr and take the min of the old and new difference
            new_minimum = max(new_arr)-min(new_arr)

            if not current_minimum:
                current_minimum = new_minimum
            else:
                current_minimum = min(new_minimum, current_minimum)
            # print(new_arr)

            # icnrement i
            i+=1
        
        return current_minimum

        
```
