# 1784. Check if Binary String Has at Most One Segment of Ones

```python
s= '011110'

# right bound
for index, value in enumerate(s[::-1]):
    if value == '1':
        right_bound = (len(s)-1)-index
        break

right_bound
```

```python
class Solution:
    def checkOnesSegment(self, s: str) -> bool:
        
        # finding the left side
        left_bound = s.index('1')

        # finding the right side
        right_bound = s.rindex('1')

        new_arr = s[left_bound:right_bound+1]

        # if the number of 1 is the same as the len of the new_arr then it is true else false
        if new_arr.count('1') == len(new_arr):
            return True

        else:
            return False
```

```python
# without using index and rindex
class Solution:
    def checkOnesSegment(self, s: str) -> bool:

        # finding the left side
        for index, value in enumerate(s):
            if value == '1':
                left_bound = index
                break

        # flip it, right bound
        for index, value in enumerate(s[::-1]):
            if value == '1':
                right_bound = (len(s)-1)-index
                break

        new_arr = s[left_bound:right_bound+1]

        # if the number of 1 is the same as the len of the new_arr then it is true else false
        if new_arr.count('1') == len(new_arr):
            return True

        else:
            return False

# time complexity of O(n)
```

```python
# answer
class Solution:
    def checkOnesSegment(self, s: str) -> bool:

        # because 'at Most One Segment of Ones' means as soon as it sees a 01 means another segment of 1 has been started since it does not have leading 0's as well
        if '01' in s:
            return False
        else:
            return True

# time complexity of O(n)
```

```python
# Input: 
s = "00"
# Output: true
 
sol = Solution()
sol.checkOnesSegment(s)
```
