# 268. Missing Number

```python
from typing import List

class Solution:
    def missingNumber(self, nums: List[int]) -> int:

        # sort first
        nums.sort()

        for index, value in enumerate(nums):
            if index!=value:
                return index

        return index+1

# time complexity of O(nlog(n)) because of the sort
# space of O(n) because sort might have internal memory
```

```python
from typing import List

class Solution:
    def missingNumber(self, nums: List[int]) -> int:

        # getting the difference in sum
        summ = sum(nums)

        supposed_sum = sum([x for x in range(len(nums)+1)])

        return supposed_sum-summ


# time complexity of O(n)
# space of O(n)
```

```python
sol = Solution()

# Input: 
nums = [1,2,3]

# Output: 0
sol.missingNumber(nums)
```
