# 190. Reverse Bits

```python
class Solution:
    def reverseBits(self, n: int) -> int:

        # remove the '0b' and flip
        flip_string = bin(n)[2:][::-1]

        # add in more zeros
        string = (32 - len(flip_string))*'0'

        flip_string = '0b' + flip_string + string

        # convert back to value
        value = int(flip_string, 2)

        return value
        
```

```python
sol = Solution()

# Input: 
n = 43261596
sol.reverseBits(n)

# Output: 964176192
```
