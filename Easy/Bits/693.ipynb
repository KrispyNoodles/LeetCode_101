# 693. Binary Number with Alternating Bits

```python
class Solution:
    def hasAlternatingBits(self, n: int) -> bool:
        
        string = bin(n)[2:]

        first_char = string[0]

        for char in string[1:]:
            if char==first_char:
                return False

            else:
                first_char=char
        
        return True

# time complexity of O(k) where k is the number of bits in n
```

```python
sol = Solution()

n = 5
sol.hasAlternatingBits(n)
```
