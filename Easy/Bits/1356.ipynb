# 1356. Sort Integers by The Number of 1 Bits

```python
value = 10

bin_string = bin(value)
print(bin_string[2:])

# counting the number of 1's
bin_string[2:].count('1')
```

```python
from collections import defaultdict
from typing import List

class Solution:
    def sortByBits(self, arr: List[int]) -> List[int]:

        # a dict to store
        bin_dict = defaultdict(list)

        # O(n)
        for value in arr:

            # sort them based on the number of '1's they have (O(k)) where k is the number of elments in the bin form
            no_of_ones = bin(value)[2:].count('1')

            # put the value into the dict
            bin_dict[no_of_ones].append(value)

        # sort by least '1' to most
        new_dict = sorted(bin_dict.items(), key=lambda x: x[0], reverse=False)

        answer = []

        # put it back into asnwer
        for key, value in new_dict:
            answer.extend(sorted(value))

        return answer

# time complexity of O(nlog(n)) where n is the number of values in the list, because of the sorting operation
```

```python
# Input: 
arr = [0,1,2,3,4,5,6,7,8]
# Output: [0,1,2,4,8,3,5,6,7]

sol = Solution()
sol.sortByBits(arr)
```

```python
# Input: 
arr = [1024,512,256,128,64,32,16,8,4,2,1]
# Output: [1,2,4,8,16,32,64,128,256,512,1024]

sol = Solution()
sol.sortByBits(arr)
```
