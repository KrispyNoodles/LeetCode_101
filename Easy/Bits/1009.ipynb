# 1009. Complement of Base 10 Integer

```python
val = '100011'
int(val, 2)
```

```python
class Solution:
    def bitwiseComplement(self, n: int) -> int:

        # get bit version
        bit_ver = bin(n)[2:]
        
        final_ans = ''

        for char in bit_ver:
            if char == '1':
                final_ans+='0'
            else:
                final_ans+='1'

        return int(final_ans, 2)
```

```python
# Input: 
n = 5
# Output: 2
# Explanation: 5 is "101" in binary, with complement "010" in binary, which is 2 in base-10.

sol = Solution()
sol.bitwiseComplement(n)
```
