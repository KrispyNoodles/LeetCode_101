# 136. Single Number

```python
from typing import List

class Solution:
    def singleNumber(self, nums: List[int]) -> int:
        
        # sort
        nums.sort()

        i=0

        while i < (len(nums)-1):

            if nums[i]==nums[i+1]:
                
                i+=2
            else:
                return nums[i]

        # it means the answer is the last value
        return nums[i]

# time complexity of O(nlog(n)) because of the sort
# space comeplxity of O(n) because the sort may allocate extra temporary memory internally
```

```python
from typing import List

class Solution:
    def singleNumber(self, nums: List[int]) -> int:
        
        answer = 0 

        for number in nums:

            answer = answer^number
        
        return answer

# we can use XOR
# when the values are the same it will give 0
# and when it is 0 and a value, it will return the value
```
