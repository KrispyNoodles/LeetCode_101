# 868. Binary Gap

```python
bin_val = '10000'


# convert into int
val = int(bin_val)

# to remove trailing 0's
while str(val)[-1] == '0':
    val = val//10

print(val)
```

```python
class Solution:
    def binaryGap(self, n: int) -> int:
        
        bin_str = bin(n)
        bin_str = bin_str[2:]

        max_gap = 0
        current_gap = 0

        # remove trailing 0's and if the len<=1 remove
        val = int(bin_str)

        # to remove trailing 0's
        while str(val)[-1] == '0':
            val = val//10

        # conver back into str
        val = str(val)

        # if len of val less than equal 1 return 0, means it is a single value or all 0
        if len(val)<=1:
            return 0

        for i in range(len(val)):
            
            if val[i] == '1':

                # update the max_gap
                max_gap = max(max_gap, current_gap)

                # refresh current_gap
                current_gap = 0

            else:
                # increment the current_gap
                current_gap+=1

        return max_gap + 1

# time complexity of O(k^2) because k is the time it takes to convert into a string and k is the max length of trailing 0's that it will have, where k is the lenght of n in binary
```

```python
# Input: 
n = 8
# Output: 0

sol = Solution()
sol.binaryGap(n)
```
