# 1266. Minimum Time Visiting All Points

```python
# point a-b, then shift
point_1 = [1,1]
point_2 = [3,4]

points = [point_1, point_2]

current_point = None
total_movement = 0

for point in points:

    # the current_point be the first point
    # then continue
    if not current_point:
        current_point = point
        continue
    
    x_diff = current_point[0] - point[0]
    y_diff = current_point[1] - point[1]

    # convert to abs
    x_diff = abs(x_diff)
    y_diff = abs(y_diff)

    # get the min
    minimum = min(x_diff,y_diff)

    total_movement = minimum + (x_diff-minimum) + (y_diff-minimum)

    # update the new current_point
    current_point = point

    print(total_movement)
```

```python
from typing import List

class Solution:
    def minTimeToVisitAllPoints(self, points: List[List[int]]) -> int:
        
        current_point = None
        total_movement = 0

        for point in points:

            # the current_point be the first point
            # then continue
            if not current_point:
                current_point = point
                continue
            
            x_diff = current_point[0] - point[0]
            y_diff = current_point[1] - point[1]

            # convert to abs
            x_diff = abs(x_diff)
            y_diff = abs(y_diff)

            # get the min
            minimum = min(x_diff,y_diff)

            # adding the difference and the previous total_movemnt
            total_movement += minimum + (x_diff-minimum) + (y_diff-minimum)

            # update the new current_point
            current_point = point

        return total_movement

# time complecity of O(n)

# no duplication of array
# space complecity of O(1)
```

```python
from typing import List

# The "Chebyshev" Logic (The max way)

class Solution:
    def minTimeToVisitAllPoints(self, points: List[List[int]]) -> int:
        
        total_movement = 0

        for point_index in range(1, len(points)):
        
            # adding the difference and the previous total_movemnt
            total_movement += max(
                abs(points[point_index][0]-points[point_index-1][0]),
                abs(points[point_index][1]-points[point_index-1][1])
                )

        return total_movement

# time complecity of O(n)

# no duplication of array
# space complecity of O(1)
```

```python
sol = Solution()

points = [[3,2],[-2,2]]
sol.minTimeToVisitAllPoints(points)
```
