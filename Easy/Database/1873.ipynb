# 1873. Calculate Special Bonus

```sql
# Write your MySQL query statement below

SELECT  e.employee_id, 
CASE
    WHEN e.employee_id%2 != 0 AND e.name NOT LIKE 'M%' THEN e.salary
    ELSE 0
END as bonus

FROM employees e
ORDER BY e.employee_id ASC


````python
import pandas as pd

data = [[2, 'Meir', 3000], [3, 'Michael', 3800], [7, 'Addilyn', 7400], [8, 'Juan', 6100], [9, 'Kannon', 7700]]
employees = pd.DataFrame(data, columns=['employee_id', 'name', 'salary']).astype({'employee_id':'int64', 'name':'object', 'salary':'int64'})
````

````python
import numpy as np

def calculate_special_bonus(employees: pd.DataFrame) -> pd.DataFrame:

    # swithc case
    condition = (
        (employees['employee_id']%2 != 0) & (~employees['name'].str.startswith('M'))
    )
    
    choices =  (
        employees['salary']
    )

    employees['bonus'] = np.where(condition, choices, 0)

    # rearramge ny employee_id by decreasing
    employees = employees.sort_values('employee_id')

    return employees[['employee_id', 'bonus']]
````

````python
calculate_special_bonus(employees)
````
